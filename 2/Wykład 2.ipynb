# Wykład 2 — Typy danych i operacje

## Wracamy do `int`

Na końcu Wykładu 1 zapowiedzieliśmy powrót do `int`. Widzieliśmy ten zapis w `int main(void)`, w funkcji `sum`, przy jej parametrach i przy zmiennych. Wtedy wystarczało nam: **`int` oznacza typ całkowity**.

Dzisiaj zapytamy dokładniej: **co typ danych mówi compilerowi — i dlaczego ma znaczenie dla wyniku programu?**

### Zaczynamy od programu, który już znamy

W W1 stopniowo budowaliśmy funkcję `sum`: najpierw wypisywała wynik, później zaczęła go zwracać. Poniżej nie ma jeszcze nowych konstrukcji.

**Zanim uruchomisz kod:**

1. Znajdź każde wystąpienie `int` i powiedz, czego dotyczy.
2. Co oznacza `void` w `main(void)`?
3. Jaki tekst pojawi się na ekranie? Co się zmieni, jeśli zamiast `sum(5, 6)` wywołamy `sum(5, 9)`?

Uruchom program, a następnie zmień **tylko jedną liczbę** i sprawdź swoją odpowiedź.

````c
#include <stdio.h>

int sum(int a, int b)
{
    return a + b;
}

int main(void)
{
    int wynik = sum(5, 6);
    printf("a+b=%d\n", wynik);
    return 0;
}
````

### Co już potrafimy wyjaśnić?

- `int a` i `int b` opisują **parametry** funkcji `sum`; przy wywołaniu przekazujemy jej argumenty `5` i `6`.
- `int sum(...)` mówi, że funkcja **zwraca wartość** typu `int`.
- `int wynik = ...` deklaruje zmienną i zapisuje w niej zwrócony wynik.
- `int main(void)` oznacza funkcję `main`, która zwraca `int` i nie przyjmuje argumentów.
- `%d` mówi funkcji `printf`, jak **wypisać** liczbę całkowitą — nie jest nazwą typu.

Tyle wystarczało nam w W1. Ale **dlaczego compiler chce znać typ `a`, `b` i `wynik`?** Co dzieje się z ich wartościami podczas obliczeń?

## Uczymy się po spirali

W dawnym materiale W2 pojawiło się pytanie: *„We had a look at functions. Was this a bad idea?”* — czy pokazanie funkcji już na pierwszym wykładzie było błędem?

**Nie.** W W1 poznaliśmy funkcje **roboczo**, ponieważ były potrzebne do napisania pierwszych programów. Nie musieliśmy wtedy znać wszystkich szczegółów `int`, `void`, parametrów ani `return`.

Tak właśnie uczymy się **po spirali**: spotykamy pojęcie w działającym programie, wyjaśniamy tyle, ile potrzeba *tu i teraz*, a później wracamy do niego z nowymi pytaniami. Obiecaliśmy, że z czasem wiele rzeczy stanie się jaśniejszych. Dzisiaj sprawdzimy tę obietnicę na **typach danych**.

## Mapa pojęć: od typu do wyniku

Typ nie jest jedynie słowem wpisanym przed nazwą zmiennej. W kolejnych eksperymentach będziemy pytać, jak wpływa na:

```text
                        TYP DANYCH
                            |
            +---------------+----------------+
            |               |                |
    reprezentację       zbiór możliwych     operacje
    i rozmiar            wartości           i konwersje
            \               |                /
                     wynik obliczenia
                            |
                    sposób wypisania
```

**Wartość, jej zapis w kodzie i sposób wyświetlenia wyniku to nie to samo.** Na przykład zapis liczby `16`, zapis `0x10` i format wypisania `%x` pełnią różne role — sprawdzimy to na programie.

Nasza droga: **`int` → rozmiar i zakres → integer arithmetic → `float`/`double` → casting → `char`/`bool`/`void` → operatory.**

## Jak będziemy pracować?

**Pytanie → przewidywanie → kod → uruchomienie → wynik lub komunikat błędu → obserwacja → wyjaśnienie → zmiana jednej rzeczy.**

Nie chcemy zapamiętywać tabel bez sprawdzenia, co naprawdę robi program. Będziemy porównywać krótkie przykłady i pytać compiler o odpowiedzi, zamiast zgadywać.

**Pytanie:** ile bajtów zajmuje `int`? Czy odpowiedź musi być identyczna na każdym komputerze? Za chwilę poznamy operator `sizeof`, dzięki któremu możemy to sprawdzić.

## Czym właściwie jest typ danych?

W W1 spotkaliśmy `int`, gdy opisywaliśmy **parametry**, **wynik funkcji** i **zmienną**. Do napisania pierwszego programu taka robocza interpretacja wystarczała.

W komputerze dane mogą być reprezentowane jako bity. Aby program mógł z nich korzystać, compiler musi znać **typ**: określa on m.in. sposób interpretacji wartości, dopuszczalne operacje i reguły konwersji. Dla typu obiektu wiąże się również z jego **reprezentacją i rozmiarem w pamięci** oraz zakresem wartości.

**Pytanie:** czy zmienna typu `int` zajmuje tyle samo miejsca co `double`? Skąd mielibyśmy to wiedzieć? Nie zgadujmy — sprawdźmy. Typy funkcji i szczególną rolę `void` omówimy osobno.

### Eksperyment 1 — pytamy o `int`

Na kartce lub w głowie postaw hipotezę: **ile bajtów potrzebuje obiekt typu `int` na naszej platformie?**

Teraz uruchom kod. Najpierw obejrzyj wynik, dopiero potem przeczytaj objaśnienie. Spróbuj zmienić **jedynie** `int` w wyrażeniu `sizeof(int)` na `char` i porównaj wyniki.

````c
#include <stdio.h>

int main(void)
{
    printf("sizeof(int) = %zu B\n", sizeof(int));
    return 0;
}
````

### Co właśnie sprawdziliśmy?

- **`sizeof` jest operatorem języka C**, a nie funkcją. Zapis `sizeof(int)` oznacza: „jaki jest rozmiar obiektu typu `int`?”.
- Wynik `sizeof` ma typ `size_t`. Do jego wypisywania w `printf` używamy formatu **`%zu`**. Nie musimy jeszcze znać szczegółów `size_t`.
- Rozmiary podstawowych typów zależą od **implementacji C i platformy docelowej**. Typowe `sizeof(int) == 4` to obserwacja na wielu współczesnych komputerach, nie gwarancja języka.
- W tym przykładzie compiler zna rozmiar typu już podczas kompilacji. Program **wyświetla rozmiar właściwy dla swojej platformy**, a nie odczytuje go dopiero z pamięci podczas działania.

**B** oznacza tutaj *bajt w rozumieniu C*. Czy bajt zawsze musi składać się z ośmiu bitów? Sprawdźmy.

### Eksperyment 2 — porównujemy rozmiary

**Przewidź**, zanim uruchomisz program:

1. Czy `sizeof(char)` da 1?
2. Czy `sizeof(int)`, `sizeof(float)` i `sizeof(double)` muszą być równe?
3. Ile **bitów** ma jeden bajt na naszej platformie?

`CHAR_BIT` z `<limits.h>` podaje liczbę bitów w bajcie C. Pozostałe rozmiary sprawdzimy przez `sizeof`.

Kod zawiera również `int liczba = 42;`, które poznaliśmy już w W1. Zobacz, co zwraca `sizeof liczba`.

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    int liczba = 42;

    printf("CHAR_BIT = %d bits per byte\n", CHAR_BIT);
    printf("sizeof(char) = %zu B\n", sizeof(char));
    printf("sizeof(int) = %zu B\n", sizeof(int));
    printf("sizeof(long long int) = %zu B\n", sizeof(long long int));
    printf("sizeof(float) = %zu B\n", sizeof(float));
    printf("sizeof(double) = %zu B\n", sizeof(double));
    printf("sizeof liczba = %zu B\n", sizeof liczba);

    return 0;
}
````

### Co wynika z porównania?

- W C **`sizeof(char)` zawsze wynosi 1** — taki jest sposób zdefiniowania jednostki „bajt”. Natomiast liczba bitów w jednym bajcie wynosi `CHAR_BIT` (na typowym współczesnym komputerze: 8).
- Pozostałych rozmiarów nie wpisujemy na stałe jako praw uniwersalnych. Wyniki zależą od implementacji.
- `sizeof(int)` pyta o **typ** (stąd nawiasy), zaś `sizeof liczba` pyta o rozmiar obiektu/wyrażenia. Nawiasy można także napisać jako `sizeof(liczba)`, ale wtedy nie są one wywołaniem funkcji.
- Jeżeli obiekt ma `n` bajtów, odpowiada mu `n * CHAR_BIT` bitów pamięci przeznaczonej na reprezentację. Sam rozmiar **nie mówi jeszcze wszystkiego** o zakresie wartości.

**Zmień tylko jedną rzecz**: `int liczba = 42;` na `double liczba = 42.0;`. Uruchom ponownie. Czy zmienia się `sizeof liczba`? Czy zmienia się też `sizeof(int)`?

Do sposobu reprezentowania `float` i `double` wrócimy dalej; teraz interesuje nas wyłącznie rozmiar.

### Co zabieramy z tego eksperymentu?

**Typ to nie dekoracja nazwy.** Wybór typu ma konsekwencje dla tego, jak compiler interpretuje wartości, ile miejsca potrzebują obiekty oraz jakie działania możemy wykonać.

Umiemy już **zmierzyć rozmiar zamiast wierzyć w tabelę**. Następny krok: utworzymy zmienne typu `int`, przypiszemy im wartości i sprawdzimy, jak ich typ wpływa na zapis i wynik obliczeń. Zakresy wartości i sytuacje błędne zbadamy w osobnych eksperymentach.

## `int` — deklaracja, inicjalizacja, przypisanie

Zmierzyliśmy już rozmiar obiektu typu `int`. Teraz zaczniemy **przechowywać i zmieniać wartości**. W W1 robiliśmy to między innymi w instrukcji `int wynik = sum(5, 6);`.

```c
int a = 90;  // deklaracja i inicjalizacja
int b;       // sama deklaracja
b = 10;      // przypisanie wartości
```

- **Deklaracja** `int b;` wprowadza nazwę i typ zmiennej.
- **Inicjalizacja** `int a = 90;` nadaje wartość przy deklaracji.
- **Przypisanie** `b = 10;` ustawia lub zmienia wartość istniejącej zmiennej; `=` nie oznacza tu sprawdzania równości.

**Pytanie:** jakie wartości wypisze program poniżej, zanim i po tym, jak zmienimy `a`? Zauważ, że `b` otrzymuje wartość **przed pierwszym odczytem**.

````c
#include <stdio.h>

int main(void)
{
    int a = 90;
    int b;
    b = 10;

    printf("a=%d, b=%d\n", a, b);
    a = 91;
    printf("after assignment: a=%d, b=%d\n", a, b);

    return 0;
}
````

### Obserwacja — typ zostaje, wartość się zmienia

`a` i `b` mają ten sam typ `int`, ale różne wartości. Instrukcja `a = 91;` zmienia **wartość zmiennej**, nie jej typ. W `printf` używamy `%d` do wypisania wartości typu `int` jako liczby dziesiętnej.

**Eksperyment:** zamień tylko `b = 10;` na `b = -10;`. Uruchom ponownie: czy zmiana znaku wymaga innego typu? Następnie przywróć `b = 10;`.

Na razie dbamy, żeby każda odczytywana zmienna miała już przypisaną wartość. Pytanie, co dzieje się **bez inicjalizacji**, zostawiamy do następnej części.

## Jedna wartość, różne zapisy w kodzie

**Pytanie:** czy zapisy `16`, `020` i `0x10` oznaczają trzy różne liczby?

W C liczby całkowite można zapisywać m.in. w systemie dziesiętnym (bez przedrostka), **ósemkowym** (początkowe `0`) i **szesnastkowym** (przedrostek `0x` lub `0X`). Tak zapisane fragmenty kodu to *literały całkowite*.

**Przewidź wynik:** jaką liczbę wypisze `printf("%d", ...)` dla każdej z trzech zmiennych?

````c
#include <stdio.h>

int main(void)
{
    int dziesietnie = 16;
    int osemkowo = 020;
    int szesnastkowo = 0x10;

    printf("%d %d %d\n", dziesietnie, osemkowo, szesnastkowo);
    return 0;
}
````

### Skąd wzięły się trzy szesnastki?

`16`, `020` i `0x10` są **różnymi zapisami tej samej wartości**. Przedrostek zmienia **sposób odczytania cyfr w kodzie źródłowym**, nie typ zmiennej ani samą liczbę.

`%d` nie „konwertuje zmiennej na liczbę dziesiętną” — sprawia, że `printf` wypisuje jej wartość **w zapisie dziesiętnym**.

**Eksperyment:** zmień tylko `0x10` na `0x11`, uruchom ponownie i wyjaśnij, dlaczego zmieniła się ostatnia wypisana liczba. Potem przywróć `0x10`.

````c
#include <stdio.h>

int main(void)
{
    int a = 20;
    int b = 020;
    int c = 0x20;

    printf("20 = %d\n", a);
    printf("020 = %d\n", b);
    printf("0x20 = %d\n", c);
    return 0;
}
````

### Te same cyfry, inne wartości

W drugim eksperymencie cyfry `20` występują w trzech systemach pozycyjnych:

- `20` to dwadzieścia (system dziesiętny),
- `020` to szesnaście (system ósemkowy),
- `0x20` to trzydzieści dwa (system szesnastkowy).

**Pytanie:** dlaczego `020` **nie** oznacza po prostu „20 z zerem na początku”?

Zapamiętaj regułę czytania literałów, a nie konkretne wyniki z tabeli.

## Ten sam obiekt, różny sposób wyświetlenia: `printf`

Znany z W1 zapis `printf("%d", a)` służy do wypisania `int` dziesiętnie. Dla **`unsigned int`**, czyli odmiany typu całkowitego bez wartości ujemnych, możemy zastosować różne specyfikatory:

- `%u` — zapis dziesiętny,
- `%o` — zapis ósemkowy,
- `%x` / `%X` — zapis szesnastkowy z małymi / wielkimi literami.

**Pytanie:** czy różne specyfikatory zmieniają wartość zmiennej `liczba`, czy tylko tekst wypisany na ekranie?

Na szczegóły zakresu `unsigned int` przyjdzie czas w kolejnej części. Na razie istotne jest, aby **typ argumentu pasował do specyfikatora** w `printf`.

````c
#include <stdio.h>

int main(void)
{
    unsigned int liczba = 16;

    printf("decimal: %u\n", liczba);
    printf("octal:   %o\n", liczba);
    printf("hex:     %x\n", liczba);
    printf("HEX:     %X\n", liczba);

    return 0;
}
````

### Wartość ≠ literał ≠ tekst wypisany przez `printf`

Porównajmy trzy rzeczy:

1. **Literał** — jak zapisaliśmy liczbę w kodzie źródłowym: `16`, `020` albo `0x10`.
2. **Wartość** — w tym przypadku zawsze ta sama, szesnaście.
3. **Format wypisania** — `%u`, `%o`, `%x` lub `%X` decyduje, jakie cyfry zobaczymy.

**Eksperyment:** zmień jedynie `unsigned int liczba = 16;` na `unsigned int liczba = 31;`. Przewidź zapisy dziesiętny, ósemkowy i szesnastkowy, a potem je porównaj.

**Uwaga:** `%d` oczekuje argumentu typu `int`, a `%u`, `%o` i `%x`/`%X` argumentu typu `unsigned int`. Nie mieszamy tych wymagań przypadkowo. W tym miejscu pokazujemy zaledwie dwa warianty typu całkowitego — pełniejsze omówienie zakresów będzie później.

### Co już umiemy?

- Zadeklarować `int`, nadać mu wartość przy deklaracji lub później przez przypisanie.
- Zmienić wartość bez zmiany typu.
- Rozpoznać zapis dziesiętny, ósemkowy i szesnastkowy literału.
- Odróżnić **wartość w programie** od jej **sposobu wyświetlania**.

**Pytanie:** jeżeli `a = 0x10` i `b = 020`, czy `a` i `b` mają tę samą wartość? Czy `printf` może wypisać obie w zapisie dziesiętnym?

Teraz zobaczymy, dlaczego poprawna inicjalizacja i ograniczony zakres typu mają znaczenie. To będzie przedmiotem W2.4.

## W2.4 — Ile mieści się w `int`?

### 1. Zmienna musi mieć wartość, zanim ją odczytamy

Wiemy już, że `int a = 90;` oznacza **deklarację i inicjalizację**, a `int b;` — tylko deklarację.

**Pytanie:** co wypisze ten fragment?

```c
int a = 90;
int b, c;
printf("a=%d, b=%d, c=%d\n", a, b, c); // b i c nie mają nadanych wartości!
```

To **błędny fragment**, nie wzór do uruchamiania: odczyt niezainicjalizowanych lokalnych `int` może prowadzić do *undefined behavior* (UB). Nawet jeśli zobaczymy zera albo inne liczby, nie możemy im ufać.

Zobaczmy poprawioną wersję.

````c
#include <stdio.h>

int main(void)
{
    int a = 90;
    int b, c;

    b = 10;
    c = 40;
    printf("a=%d, b=%d, c=%d\n", a, b, c);

    return 0;
}
````

**Obserwacja:** `b` i `c` są odczytywane **dopiero po przypisaniu wartości**. Otrzymujemy `a=90, b=10, c=40`.

**Eksperyment:** zamień `int b, c;` oraz dwa przypisania na jedną instrukcję `int b = 10, c = 40;`. Czy wynik się zmieni?

Wróć do oryginalnego kodu przed następnym przykładem.

### 2. Jakie wartości może przechowywać `int`?

W W2.2 sprawdziliśmy, ile bajtów zajmuje `int`. Teraz zapytamy o jego **zakres**.

Nagłówek `<limits.h>` podaje granice dla bieżącej implementacji C:

- `INT_MIN` — najmniejsza wartość `int`,
- `INT_MAX` — największa wartość `int`,
- `UINT_MAX` — największa wartość `unsigned int`.

**Pytanie:** jakie liczby otrzymamy na komputerze, na którym właśnie pracujemy?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    printf("INT_MIN = %d\n", INT_MIN);
    printf("INT_MAX = %d\n", INT_MAX);
    printf("UINT_MAX = %u\n", UINT_MAX);

    return 0;
}
````

Na `dodo` odczytaliśmy `INT_MIN = -2147483648`, `INT_MAX = 2147483647` oraz `UINT_MAX = 4294967295`. To **wyniki dla tej platformy**, nie liczby gwarantowane dla każdej implementacji C.

Wiemy też z W2.2, że na `dodo` `sizeof(int) = 4` i `CHAR_BIT = 8`, więc obiekt `int` zajmuje **32 bity**.

**Pytanie:** skąd właśnie taki zakres i dlaczego `INT_MIN` ma wartość o jeden mniejszą co do modułu niż `INT_MAX`? Zajrzyjmy do reprezentacji bitowej.

### Jak zapisać liczbę w bitach? Model 8-bitowy

Każdy **bit** może przyjąć wartość `0` albo `1`. Osiem bitów daje `2^8 = 256` różnych wzorców.

Dla wygody najpierw przyjrzymy się **8-bitowemu modelowi liczb ze znakiem w kodzie uzupełnień do dwóch** (*two's complement*). To model do nauki — **nie** twierdzimy, że typ `int` w C zajmuje 8 bitów.

| Zapis binarny (8 bitów) | Wartość |
|:------------------------|--------:|
| `0000 0000` | 0 |
| `0000 0001` | 1 |
| `0111 1111` | 127 |
| `1000 0000` | −128 |
| `1000 0001` | −127 |
| `1111 1110` | −2 |
| `1111 1111` | −1 |

W takiej reprezentacji **najstarszy bit** (po lewej stronie zapisu binarnego) wskazuje, czy liczba jest nieujemna (`0`), czy ujemna (`1`). Nie jest jednak osobną „lampką znaku”: **uczestniczy w kodowaniu całej wartości**. Dla 8 bitów ma wagę `−128`.

**Pytanie:** dlaczego największa dodatnia wartość to `127`, a najmniejsza ujemna to `−128`?

**Odpowiedź:** W 8-bitowym kodzie uzupełnień do dwóch najstarszy bit ma wagę **−128**, a pozostałe siedem bitów ma dodatnie wagi **64, 32, 16, 8, 4, 2 i 1**.

- **Największa dodatnia wartość:** `0111 1111` = 64 + 32 + 16 + 8 + 4 + 2 + 1 = **127**. Najstarszy bit jest zerem, więc nie wnosi wartości −128.
- **Najmniejsza ujemna wartość:** `1000 0000` = −128 + 0 = **−128**. Wszystkie pozostałe bity są zerami.
- Pozostałe przykłady z tabeli: `1000 0001` = −128 + 1 = **−127** oraz `1111 1110` = −128 + 126 = **−2**.

Mamy zatem **128 wzorców nieujemnych** (od 0 do 127) i **128 ujemnych** (od −128 do −1). Zero należy do pierwszej grupy — stąd jeden dodatkowy wynik ujemny i brak dodatniego `+128` w tym 8-bitowym modelu.

### Jak wygląda nasz 32-bitowy `int`?

Na `dodo` obiekt `int` zajmuje **4 bajty po 8 bitów**. Gdy jest przechowywany w pamięci, mieści się więc w nim 32-bitowy wzorzec. GCC na tej platformie używa dla `int` **uzupełnień do dwóch**.

| Wartość typu `int` | Wzorzec 32 bitów |
|:---------------------|:-----------------|
| `0` | `00000000 00000000 00000000 00000000` |
| `INT_MAX = 2147483647` | `01111111 11111111 11111111 11111111` |
| `INT_MIN = -2147483648` | `10000000 00000000 00000000 00000000` |
| `-1` | `11111111 11111111 11111111 11111111` |

**Zwróć uwagę na pierwszy bit od lewej**: dla `INT_MAX` jest równy `0`, a dla `INT_MIN` — `1`. W 32-bitowym kodzie uzupełnień do dwóch jego waga wynosi `−2^31`.

Odstępy co osiem bitów są tu **wyłącznie dla czytelności**. Tabela przedstawia **logiczny zapis liczby**, a nie kolejność adresów bajtów w pamięci. Kolejnością bajtów (*endianness*) zajmiemy się później.

### Te same bity — dlaczego typ ma znaczenie?

Na naszej 32-bitowej platformie wzorzec:

```text
10000000 00000000 00000000 00000000
```

odczytujemy jako **`−2147483648`**, gdy interpretujemy go jako `int`, ale jako **`2147483648`**, gdy interpretujemy go jako `unsigned int`.

Podobnie wzorzec złożony z 32 jedynek odpowiada **`−1`** jako `int`, a **`4294967295`** jako `unsigned int`.

To pokazuje różnicę między **wzorcem bitów** a **wartością wynikającą z typu**. Nie należy z tego wnioskować, że każde przepełnienie `int` musi dawać określony wzorzec: standard C nie gwarantuje takiego wyniku.

**Pytanie:** jeśli zaczynamy od `INT_MAX`, którego zapis binarny to `0` i 31 jedynek, jaki 32-bitowy wzorzec mogłoby dać sprzętowe dodanie jedynki? Teraz sprawdzimy, co rzeczywiście wypisze program na `dodo`.

### 3. Przepełnienie `int` — sprawdzamy eksperymentalnie

To **celowy eksperyment z niepoprawną operacją**. W standardowym C, gdy suma liczb typu `int` przekracza dopuszczalny zakres, następuje **undefined behavior**.

Nie znamy więc wyniku zagwarantowanego przez język. Ale nadal możemy **uruchomić program i zobaczyć, co zrobi ta konkretna kompilacja**.

**Pytanie:** czy po dodaniu `1` zobaczymy liczbę ujemną, błąd, czy może coś innego?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    int a = INT_MAX;

    printf("przed=%d\n", a);
    a = a + 1;  // EKSPERYMENT: signed overflow to undefined behavior w C!
    printf("po=%d\n", a);

    return 0;
}
````

### Co zaobserwowaliśmy, a co mówi standard?

Na `dodo` uruchomiliśmy program z przepełnieniem `int` i otrzymaliśmy:

```text
przed=2147483647
po=-2147483648
```

Widzieliśmy przed chwilą, że dla naszej **32-bitowej reprezentacji uzupełnień do dwóch** odpowiada to zmianie wzorca:

```text
  01111111 11111111 11111111 11111111   (INT_MAX)
+ 00000000 00000000 00000000 00000001   (+1)
---------------------------------------
  10000000 00000000 00000000 00000000   (INT_MIN)
```

Taki wynik można zrozumieć, patrząc na działanie 32-bitowego dodawania maszynowego i interpretację otrzymanych bitów. **To wyjaśnia obserwację, ale nie jest regułą standardu C.** Dla przepełnienia `int` standard mówi: **undefined behavior**.

Wynik zależy od kodu wygenerowanego przez kompilator i od warunków uruchomienia. Przy innych ustawieniach optymalizacji może być inny — choć nie musi. Sam wynik nie pozwala wskazać konkretnej optymalizacji.

**Wniosek:** najpierw widzimy, **co wykonał nasz program**; osobno ustalamy, **co gwarantuje język C**.

### 4. A co z `unsigned int`?

`unsigned int` nie przechowuje wartości ujemnych. Tutaj C **gwarantuje** arytmetykę modulo `UINT_MAX + 1`.

**Wyjaśnienie `1u`:** sufiks `u` oznacza literał typu `unsigned int` (w przeciwieństwie do `1`, które ma typ `int`). W tym przykładzie również `a + 1` da ten sam rezultat, ponieważ przy dodawaniu do `unsigned int a` jedynka zostanie przekonwertowana na `unsigned int`. Do zasad konwersji wrócimy w W2.7.

**Pytanie:** co wypisze bardzo podobny program, jeśli zaczniemy od `UINT_MAX` i dodamy `1u`?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    unsigned int a = UINT_MAX;

    printf("przed=%u\n", a);
    a = a + 1u;
    printf("po=%u\n", a);

    return 0;
}
````

**Obserwacja:** po przekroczeniu `UINT_MAX` otrzymujemy `0`. Tym razem **nie jest to przypadek UB** — takie zachowanie gwarantuje C.

Porównaj dwa eksperymenty:

- `int`: widzieliśmy określony wynik na konkretnej maszynie, ale **język go nie gwarantuje**;
- `unsigned int`: również zobaczyliśmy zawinięcie, ale **język gwarantuje** wynik modulo zakresu.

**Eksperyment:** zmień `unsigned int a = UINT_MAX;` na `unsigned int a = UINT_MAX - 1u;`. Co powinno się wypisać po dodaniu `1u`?

### Co wynika z W2.4?

- Wartość zmiennej lokalnej należy ustalić **zanim ją odczytamy**.
- Zakresy `int` i `unsigned int` można sprawdzić przez `<limits.h>`.
- Przepełnienie `int` to **undefined behavior** — eksperymentalny wynik nie jest gwarancją standardu.
- Dla `unsigned int` wynik zawija się modulo zakresu **zgodnie ze standardem**.

**Pytanie:** skoro `int` przechowuje liczby całkowite, to co będzie wynikiem `5 / 6`? W W2.5 przyjrzymy się działaniom arytmetycznym na takich liczbach.

## W2.5 — Działania na liczbach całkowitych

### Jeszcze o bitach: reprezentacja `unsigned int`

W W2.4 oglądaliśmy **8-bitowy model liczb ze znakiem** w kodzie uzupełnień do dwóch. Jak te same bity odczytamy jako **liczbę bez znaku**?

Dla **modelu 8-bitowego bez znaku** wszystkie bity mają dodatnie wagi: **128, 64, 32, 16, 8, 4, 2, 1**.

| Zapis binarny (8 bitów) | Wartość bez znaku |
|:-----------------------|------------------:|
| `0000 0000` | 0 |
| `0000 0001` | 1 |
| `0111 1111` | 127 |
| `1000 0000` | 128 |
| `1111 1111` | 255 |

**Pytanie:** dlaczego `1000 0000` w modelu ze znakiem oznaczało **−128**, a tutaj oznacza **128**? Bity się nie zmieniły — zmieniła się ich **interpretacja**.

Na `dodo` `unsigned int` zajmuje 32 bity, zatem jego zakres to **od 0 do `2^32 − 1 = 4294967295`**, a najstarszy bit ma wagę **`2^31`**. Wzorzec 32 jedynek oznacza tutaj `UINT_MAX`. To konkretny przypadek implementacji na `dodo`, nie wymóg, aby wszędzie `unsigned int` miał 32 bity.

W W2.4 zauważyliśmy, że po `UINT_MAX + 1u` otrzymujemy 0. W C dla liczb bez znaku jest to **gwarantowane** działanie modulo liczby wszystkich możliwych wartości.

### Dodawanie, odejmowanie i mnożenie

Dotychczas przechowywaliśmy i wypisywaliśmy liczby. Teraz wykonamy działania:

- `+` — dodawanie;
- `-` — odejmowanie;
- `*` — mnożenie.

**Pytanie:** jeśli `a = 5` i `b = 4`, jakie wyniki wypisze program?

````c
#include <stdio.h>
int main(void)
{
    int a = 5, b = 4;

    printf("a=%d, b=%d\n", a, b);
    printf("a+b=%d\n", a + b);
    printf("a-b=%d\n", a - b);
    printf("a*b=%d\n", a * b);

    return 0;
}
````

**Obserwacja:** otrzymujemy `9`, `1` i `20`. Wszystkie działania wykonujemy na `int` i wypisujemy wyniki za pomocą `%d`.

**Eksperyment:** zmień tylko `b = 4` na `b = 6`. Przewidź wyniki i porównaj z programem.

Pamiętajmy o W2.4: przy arytmetyce `int` matematyczny wynik nie może przekroczyć dopuszczalnego zakresu, jeżeli oczekujemy zachowania zdefiniowanego przez C.

### Dzielenie `/` i reszta z dzielenia `%`

**Pytanie:** jaki wynik da `5 / 6`, jeśli obie liczby są typu `int`?

W C dzielenie **dwóch liczb całkowitych** daje wynik całkowity: część ułamkowa jest **odrzucana w kierunku zera**. Zatem `5 / 6` to `0`, a `6 / 5` to `1`.

Operator `%` pozwala obliczyć **resztę z dzielenia**, np. `6 % 5` daje `1`. Aby wydrukować dosłowny znak procenta wewnątrz napisu `printf`, piszemy `%%`.

Przewidź cztery wyniki, zanim uruchomisz kod.

````c
#include <stdio.h>
int main(void)
{
    int a = 5, b = 6;

    printf("a/b=%d\n", a / b);
    printf("b/a=%d\n", b / a);
    printf("a%%b=%d\n", a % b);
    printf("b%%a=%d\n", b % a);

    return 0;
}
````

**Obserwacja:** `5 / 6 = 0`, `6 / 5 = 1`, `5 % 6 = 5` i `6 % 5 = 1`.

Dzielenie całkowite **nie zaokrągla do najbliższej liczby**. Dla dodatnich argumentów po prostu odrzuca część ułamkową.

**Eksperyment:** zmień `a = 5` na `a = 13`. Co otrzymasz dla `a / b` i `a % b`?

**Uwaga:** dzielenie przez `0` i obliczanie reszty z dzielenia przez `0` są w C błędnymi operacjami (*undefined behavior*). Nie eksperymentujemy z nimi tutaj.

### A co z liczbami ujemnymi?

W C dzielenie całkowite odrzuca część ułamkową **w kierunku zera**, także dla liczb ujemnych. Dlatego `−7 / 3` daje `−2`, a nie `−3`.

Reszta spełnia zależność `a = (a / b) * b + (a % b)`, gdy dzielenie jest zdefiniowane. Jeśli reszta nie jest zerem, ma **znak dzielnej** (pierwszej liczby).

**Pytanie:** jakie wyniki otrzymamy dla `−7 % 3` i `7 % −3`?

````c
#include <stdio.h>
int main(void)
{
    int a = -7, b = 3;

    printf("%d/%d=%d\n", a, b, a / b);
    printf("%d%%%d=%d\n", a, b, a % b);
    printf("%d/%d=%d\n", -a, -b, (-a) / (-b));
    printf("%d%%%d=%d\n", -a, -b, (-a) % (-b));

    return 0;
}
````

**Obserwacja:** `−7 / 3 = −2`, `−7 % 3 = −1`, `7 / −3 = −2` i `7 % −3 = 1`.

Sprawdźmy: `−7 = (−2) * 3 + (−1)`. Wartość `−1` jest właśnie resztą z dzielenia `−7` przez `3`.

**Pytanie:** skoro `5 / 6` wykonane na `int` daje `0`, jak uzyskać wynik z częścią ułamkową?

Do tego potrzebujemy typów **`float`** i **`double`**. W W2.6 pokażemy nie tylko ich działanie, ale również **reprezentację bitową** (znak, wykładnik i część ułamkową, na typowej platformie według IEEE 754) oraz związane z nią ograniczenia dokładności.

## W2.6 — `float` i `double`: liczby z częścią ułamkową

W W2.5 stwierdziliśmy, że `5 / 6` daje `0`, gdy oba argumenty są typu `int`. Jak otrzymać przybliżoną wartość **0,8333…**?

Do zapisywania liczb z częścią ułamkową używamy m.in. typów **`float`** i **`double`**.

- `5.0` jest w C literałem typu **`double`**.
- `5.0f` jest literałem typu **`float`**; sufiks `f` wskazuje typ, tak jak poznane wcześniej `u` w `1u`.

**Pytanie:** co wypisze ten sam iloraz obliczony na trzech typach?

````c
#include <stdio.h>

int main(void)
{
    int ai = 5, bi = 6;
    float af = 5.0f, bf = 6.0f;
    double ad = 5.0, bd = 6.0;

    printf("int:    %d\n", ai / bi);
    printf("float:  %.6f\n", af / bf);
    printf("double: %.6f\n", ad / bd);

    return 0;
}
````

**Obserwacja:** `int` daje `0`, natomiast `float` i `double` wypisują `0.833333` — z dokładnością do sześciu miejsc po przecinku.

**Ważne:** taki sam tekst wypisany na ekranie nie oznacza jeszcze, że obie wartości zmiennoprzecinkowe przechowano identycznie.

W `printf` format **`%.6f`** oznacza sześć cyfr po kropce; `%e` wykorzystamy później do zapisu naukowego. Do `printf` możemy stosować **`%f` zarówno dla `float`, jak i `double`**: argument typu `float` jest w takim wywołaniu automatycznie przekazywany jako `double`. `%lf` w `printf` (od C99) oznacza to samo co `%f`; nie należy tej reguły przenosić na `scanf`.

**Eksperyment:** zmień w pierwszej parze `ai = 5` na `ai = 12`. Czy pozostałe dwa wyniki się zmienią?

### Ile pamięci zajmują `float` i `double`?

W W2.2 poznaliśmy `sizeof`. Teraz sprawdźmy, **ile bitów dokładności** otrzymujemy w typach zmiennoprzecinkowych.

Nagłówek `<float.h>` udostępnia m.in. `FLT_RADIX`, `FLT_MANT_DIG`, `DBL_MANT_DIG`, `FLT_DIG` i `DBL_DIG`. Nie musimy jeszcze znać szczegółów tych nazw — odczytajmy wyniki.

**Pytanie:** czy dwa razy większa liczba bajtów oznacza po prostu „dwa razy więcej cyfr po przecinku”?

````c
#include <stdio.h>
#include <float.h>

int main(void)
{
    printf("sizeof(float)  = %zu B\n", sizeof(float));
    printf("sizeof(double) = %zu B\n", sizeof(double));
    printf("FLT_RADIX = %d\n", FLT_RADIX);
    printf("FLT_MANT_DIG = %d\n", FLT_MANT_DIG);
    printf("DBL_MANT_DIG = %d\n", DBL_MANT_DIG);
    printf("FLT_DIG = %d\n", FLT_DIG);
    printf("DBL_DIG = %d\n", DBL_DIG);

    return 0;
}
````

### Jak zapisujemy wartości zmiennoprzecinkowe w bitach?

Na typowej współczesnej platformie, również na `dodo`, `float` i `double` korzystają z formatów **IEEE 754**: odpowiednio **binary32** i **binary64**.

| Pole reprezentacji | `float` (binary32) | `double` (binary64) |
|:-------------------|--------------------:|---------------------:|
| **Znak** | 1 bit | 1 bit |
| **Wykładnik** | 8 bitów | 11 bitów |
| **Część ułamkowa** | 23 bity | 52 bity |
| **Łącznie** | **32 bity** | **64 bity** |

```
float:   [znak: 1] [wykładnik: 8]  [część ułamkowa: 23]
double:  [znak: 1] [wykładnik: 11] [część ułamkowa: 52]
```

Liczba nie jest tu przechowywana tak jak zwykły `int`. W przybliżeniu myślimy o niej jak o **zapisie naukowym w systemie dwójkowym**: znak, znaczące cyfry i potęga dwójki. Dla zwykłych, niezerowych liczb *znormalizowanych* znaczące cyfry mają na początku dodatkową, **domyślną jedynkę**, której nie zapisujemy w 23 lub 52 bitach części ułamkowej.

Standard C **nie wymaga**, aby każda implementacja używała dokładnie tych formatów. W dalszych przykładach wyjaśniamy reprezentację właściwą dla naszej platformy. Zera, bardzo małe liczby oraz nieskończoności mają szczególne zapisy — na razie odkładamy szczegóły.

### Przykład: jak zakodować `1.0` i `0.5`?

Spójrzmy na dwie liczby, które **można dokładnie zapisać binarnie**:

```
1.0 = 1.0 (binarnie) × 2^0
0.5 = 1.0 (binarnie) × 2^(-1)
```

Dla zwykłych liczb znormalizowanych pole wykładnika nie zawiera bezpośrednio `0` albo `−1`: do właściwego wykładnika dodaje się **przesunięcie** (*bias*) — 127 dla `float`, 1023 dla `double`.

| Wartość | Znak | Wykładnik `float` (8 bitów) | Wykładnik `double` (11 bitów) | Część ułamkowa |
|:--------|:----:|:---------------------------:|:-----------------------------:|:----------------|
| `1.0` | `0` | `01111111` (127) | `01111111111` (1023) | same zera |
| `0.5` | `0` | `01111110` (126) | `01111111110` (1022) | same zera |

Dla obu wartości **ukryta jedynka** wraz z zerami w części ułamkowej daje znaczące cyfry `1.0`. Zmiana wykładnika z 0 na −1 dzieli wartość przez dwa.

**Pytanie:** jak zmienić `1.0` w `−1.0` bez zmiany jej wielkości? W tych formatach wystarczy zmienić **bit znaku** z `0` na `1`.

To logiczny zapis pól liczby, **nie** kolejność bajtów pod adresami pamięci.

### Czy każda liczba dziesiętna ma dokładny zapis binarny?

**Pytanie:** dlaczego `0.5` można zapisać dokładnie, ale z `0.1` jest trudniej?

`0.5 = 1/2`, więc w systemie dwójkowym ma skończony zapis: `0.1₂`. Natomiast `0.1` w systemie dziesiętnym ma w systemie dwójkowym **nieskończone rozwinięcie okresowe**:

```
0.1 (dziesiętnie) ≈ 0.00011001100110011... (binarnie)
```

Typy `float` i `double` mogą przechować tylko skończoną liczbę bitów, więc zapisują **przybliżenia** tej wartości.

Przewidź, czy oba typy wypiszą dokładnie `0.1`, gdy poprosimy o większą liczbę cyfr po przecinku.

````c
#include <stdio.h>

int main(void)
{
    float a = 0.1f;
    double b = 0.1;

    printf("float:  %.9f\n", a);
    printf("double: %.17f\n", b);

    return 0;
}
````

**Obserwacja na `dodo`:**

```
float:  0.100000001
double: 0.10000000000000001
```

Program wypisuje dziesiętne rozwinięcia **przybliżeń przechowywanych w pamięci**. To nie oznacza, że dodaliśmy do `0.1` jakieś wartości w kodzie.

`float` ma mniej bitów precyzji, więc jego przybliżenie jest tu bardziej oddalone od dokładnej liczby dziesiętnej `0.1` niż przybliżenie zapisane w `double`.

Format `%.9f` wypisuje **9 cyfr po kropce**, a `%.17f` — **17 cyfr po kropce**. Liczba miejsc po kropce nie jest tym samym co liczba poprawnych **cyfr znaczących**.

**Eksperyment:** zmień `0.1f` na `0.5f` oraz `0.1` na `0.5`. Czy teraz otrzymasz dokładne połówki?

### Dlaczego `double` ma większą precyzję?

W typowych formatach IEEE 754 liczby znormalizowane mają:

- `float`: **24 bity cyfr znaczących** (23 zapisane i 1 ukryty) — około **7 cyfr dziesiętnych**;
- `double`: **53 bity cyfr znaczących** (52 zapisane i 1 ukryty) — około **15–16 cyfr dziesiętnych**.

To **cyfry znaczące**, a nie stała liczba miejsc po przecinku! Wartość `FLT_DIG` i `DBL_DIG` z `<float.h>` pomaga określić gwarantowaną precyzję dziesiętną.

**Pytanie:** czy dodanie `1` do bardzo dużej liczby zawsze zmieni wartość typu `float`? Sprawdźmy na dokładnie reprezentowanej liczbie `2^24 = 16777216`.

````c
#include <stdio.h>

int main(void)
{
    float a = 16777216.0f;
    float a1 = a + 1.0f;
    double b = 16777216.0;
    double b1 = b + 1.0;

    printf("float:  %.0f -> %.0f\n", a, a1);
    printf("double: %.0f -> %.0f\n", b, b1);

    return 0;
}
````

**Obserwacja na `dodo`:**

```
float:  16777216 -> 16777216
double: 16777216 -> 16777217
```

To **nie jest przepełnienie zakresu**, lecz **utrata precyzji**: w tym rejonie wartości kolejne liczby reprezentowalne przez `float` są oddalone od siebie bardziej niż o 1. Po dodaniu jedynki i zaokrągleniu do `float` wartość pozostaje niezmieniona.

`double` ma więcej bitów znaczących i potrafi w tym przypadku odróżnić obie liczby.

**Eksperyment:** zmień `16777216.0f` na `100.0f` oraz `16777216.0` na `100.0`. Czy w obu typach dodanie jedynki zostanie teraz zauważone?

### Działania na `double` i notacja naukowa

Na liczbach zmiennoprzecinkowych możemy wykonywać znane operacje: `+`, `-`, `*` i `/`.

W `printf` można korzystać z formatu `%e`, czyli **notacji naukowej**. Podobnie w kodzie źródłowym zapis `6.02214076e23` oznacza liczbę około `6.02214076 × 10^23` — `e23` jest częścią **zapisu dziesiętnego** literału, a nie polem wykładnika IEEE 754 w pamięci.

**Pytanie:** czy `4.5 / 1.5` da wynik `3`, czy `3.0`? Jak pokaże go `printf("%.2f", ...)`?

````c
#include <stdio.h>

int main(void)
{
    double a = 4.5, b = 1.5;

    printf("a+b=%.2f\n", a + b);
    printf("a-b=%.2f\n", a - b);
    printf("a*b=%.2f\n", a * b);
    printf("a/b=%.2f\n", a / b);
    printf("scientific: %e\n", 6.02214076e23);

    return 0;
}
````

### Co zabieramy z W2.6?

- `5 / 6` dla `int` daje `0`, ale `5.0 / 6.0` daje wynik z częścią ułamkową.
- Typowe `float` i `double` korzystają z formatu złożonego z **bitu znaku, wykładnika i części ułamkowej**, a nie z zapisu całkowitego jak `int`.
- Nie każdą liczbę dziesiętną da się zapisać dokładnie w skończonej reprezentacji binarnej.
- `double` zwykle zachowuje więcej **cyfr znaczących** niż `float`.
- Zapis `0.1f` wybiera `float`, a `0.1` bez sufiksu jest literałem `double`.
- `%.6f` i `%e` zmieniają **sposób wyświetlenia**, a nie typ przechowywanej zmiennej.

**Pytanie:** co zrobi C, jeśli w jednym wyrażeniu spotkają się różne typy, na przykład `int` i `double`? W W2.7 zajmiemy się przypisaniem, konwersjami i utratą informacji.

## Data casting
Now that we know two data types, we can try to change the way data is treated by means of casting.

Consider the following:

We can use *explicit* data casting. This is done by specifying the type to which the result should be *cast*.  
You do this by: **(type)value**.

````c
#include <stdio.h>

int main()
{
    int a = 1, b=2;
    
    printf( "%d / %d = %d\n", a, b, a/b); // we loose data here
    printf( "%d %d %lf\n", a, b, (double)a/b); // explicit cast a into a double
    
    printf( "%d\n", 1/2); // this is wrong to do with %lf!
    printf( "%lf\n", 1./2);
}
````

What happens there?  
1) So in line 7 we try to print a result of division of two ints. Since both operands are of type int the result is assumed to be of the same type. The operation cuts out any decimal part and only the integral part is left. That is why we see a zero.  
2) In line 8 we perform an explicit cast by adding (double) to one of the variables (we write (double)a/b) this informs the complier that the result should be treated as a double. This way we see a proper result.  
3) In lines 9 and 10 we print the result of 1/2, please note that adding a '.' changes the type from int to double! so 1 - an integer, but 1. - a double!

## Data assigment, data loss and casting

Mixing of types should be, if possible avoided. If necessary it should be done with care.  
Here we interpret a `double` as an `int` and than back again. As a result we suffer data loss since the output vale is not what it originally was. It is possible that your compiler will complain with a warning.

````c
#include <stdio.h>

int main(){
    double x1 = 6.28;
    int a = 2;
    printf("%d %lf \n", a, x1);
    
    a = x1; // loss of data since a = 6!
    printf("%d %lf \n", a, x1);
    
    x1 = a;
    printf("%d %lf \n", a, x1);
}
````

And here we do the same, but "more consciously", i.e we perform a data cast we have seen earlier.

````c
#include <stdio.h>

int main()
{
    double x1 = 6.28;
    int a = 2;
    
    a = (int) x1; // loss of data , but no warning
    printf("%d %lf \n", a, x1);
    
    x1 = (double)a; // x1 is not zero since I use a cast
    printf("%d %lf \n", a, x1);
}
````

## Increment / decrement
C offers some operators that are good to know. Incrementation ++ and decremantation -- operators increase or decrease *int* value by 1. Consider the following:

````c
#include <stdio.h>

int main(){
    int a = 1;
    printf("%d \n", a);
    ++a; // a = a + 1;
    printf("%d \n", a);
    ++a;
    printf("%d \n", a);
    ++a;
    printf("%d \n", a);
    a++;
    printf("%d \n", a);
    a++;
    printf("%d \n", a);
    
    a--;
    printf("%d \n", a);
    a--;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
    --a;
    printf("%d \n", a);
}
````

and:

````c
#include <stdio.h>

int main(){
    int a = 1;
    printf("%d \n", a);
    a++;
    printf("%d \n", a);
    a++;
    printf("%d \n", a);
    
    a--;
    printf("%d \n", a);
    a--;
    printf("%d \n", a);
}
````

There are two versions of those operators and apparently the two do the same. But there is a difference. The ++a is called prefix and a++ the postfix. The first is performed first and than passed for possible assigment. The second is first passed for assigment and only than incremented. Consider the two examples and note the resulting values:

````c
#include <stdio.h>

int main()
{
    int a = 1;
    int b = ++a; // prefix: first increment than copy
    printf("%d %d\n", a, b);
}
````

````c
#include <stdio.h>

int main(){
    int a = 1;
    int b = a+1; // prefix: first increment than copy
    ++a;
    printf("%d %d\n", a, b);
}
````

````c
#include <stdio.h>

int main(){
    int a = 1;
    int b = a++; // postfix: first copy than increment
    printf("a=%d b=%d\n", a, b);
}
````

We note that it is better to use the prefix version since it only invokes coping the value, while the postfix requires first to create a copy that is passed and than the incrementation, which in case of composite data types can be expensive.

## Compound Assignment Operators $+=$, $-=$, $*=$, $/=$
Another type of operators are the compound operators. Those are used to replace operation with the assigment. Such as a = a + b. Consider the following:

````c
#include <stdio.h>

int main()
{
    int a = 1;
    int b = 2;
    a += b; // a = a + b;
    printf("a=%d b=%d\n", a, b);
    a -= b; // a = a - b;
    printf("a=%d b=%d\n", a, b);
    a *= b; // a = a * b;
    printf("a=%d b=%d\n", a, b);
    a /= b; // a = a / b;
    printf("a=%d b=%d\n", a, b);
}
````

````c
#include <stdio.h>

int main(){
    double a = 3.14;
    double b = 2.73;
    a *= b; // a = a * b;
    printf("%lf %lf\n", a, b);
    a /= b; // a = a / b;
    printf("%lf %lf\n", a, b);
    a += b; // a = a + b;
    printf("%lf %lf\n", a, b);
    a -= b; // a = a - b;
    printf("%lf %lf\n", a, b);
}
````
