## Wracamy do `int`

Na końcu Wykładu 1 zapowiedzieliśmy powrót do `int`. Widzieliśmy ten zapis w `int main(void)`, w funkcji `sum`, przy jej parametrach i przy zmiennych. Wtedy wystarczało nam: **`int` oznacza typ całkowity**.

### Zaczynamy od programu, który już znamy

W W1 stopniowo budowaliśmy funkcję `sum`: najpierw wypisywała wynik, później zaczęła go zwracać. Poniżej nie ma jeszcze nowych konstrukcji.

**Zanim uruchomisz kod:**

1. Znajdź każde wystąpienie `int` i powiedz, czego dotyczy.
2. Co oznacza `void` w `main(void)`?
3. Jaki tekst pojawi się na ekranie? Co się zmieni, jeśli zamiast `sum(5, 6)` wywołamy `sum(5, 9)`?

Uruchom program, a następnie zmień **tylko jedną liczbę** i sprawdź swoją odpowiedź.

````c
#include <stdio.h>

int sum(int a, int b)
{
    return a + b;
}

int main(void)
{
    int wynik = sum(5, 6);
    printf("a+b=%d\n", wynik);
    return 0;
}
````

### Co już potrafimy wyjaśnić?

- `int a` i `int b` opisują **parametry** funkcji `sum`; przy wywołaniu przekazujemy jej argumenty `5` i `6`.
- `int sum(...)` mówi, że funkcja **zwraca wartość** typu `int`.
- `int wynik = ...` deklaruje zmienną i zapisuje w niej zwrócony wynik.
- `int main(void)` oznacza funkcję `main`, która zwraca `int` i nie przyjmuje argumentów.
- `%d` mówi funkcji `printf`, jak **wypisać** liczbę całkowitą — nie jest nazwą typu.

Tyle wystarczało nam w W1. Ale **dlaczego compiler chce znać typ `a`, `b` i `wynik`?** Co dzieje się z ich wartościami podczas obliczeń?

## Mapa pojęć: od typu do wyniku

Typ nie jest jedynie słowem wpisanym przed nazwą zmiennej. W kolejnych eksperymentach będziemy pytać, jak wpływa na:

```
                        TYP DANYCH
                            |
            +---------------+----------------+
            |               |                |
    reprezentację       zbiór możliwych     operacje
    i rozmiar            wartości           i konwersje
            \               |                /
                     wynik obliczenia
                            |

```

**Wartość, jej zapis w kodzie i sposób wyświetlenia wyniku to nie to samo.** Na przykład zapis liczby `16`, zapis `0x10` i format wypisania `%x` pełnią różne role — sprawdzimy to na programie.

Nasza droga: **`int` → rozmiar i zakres → integer arithmetic → `float`/`double` → casting → `char`/`bool`/`void` → operatory.**

## Czym właściwie jest typ danych?

W W1 spotkaliśmy `int`, gdy opisywaliśmy **parametry**, **wynik funkcji** i **zmienną**. Do napisania pierwszego programu taka robocza interpretacja wystarczała.

W komputerze dane mogą być reprezentowane jako bity. Aby program mógł z nich korzystać, compiler musi znać **typ**: określa on m.in. sposób interpretacji wartości, dopuszczalne operacje i reguły konwersji. Dla typu obiektu wiąże się również z jego **reprezentacją i rozmiarem w pamięci** oraz zakresem wartości.

**Pytanie:** czy zmienna typu `int` zajmuje tyle samo miejsca co `double`? Skąd mielibyśmy to wiedzieć? Nie zgadujmy — sprawdźmy. Typy funkcji i szczególną rolę `void` omówimy osobno.

### Eksperyment 1 — pytamy o `int`

Na kartce lub w głowie postaw hipotezę: **ile bajtów potrzebuje obiekt typu `int` na naszej platformie?**

Teraz uruchom kod. Najpierw obejrzyj wynik, dopiero potem przeczytaj objaśnienie. Spróbuj zmienić **jedynie** `int` w wyrażeniu `sizeof(int)` na `char` i porównaj wyniki.

````c
#include <stdio.h>

int main(void)
{
    printf("sizeof(int) = %zu B\n", sizeof(int));
    return 0;
}
````

### Co właśnie sprawdziliśmy?

- **`sizeof` jest operatorem języka C**, a nie funkcją. Zapis `sizeof(int)` oznacza: „jaki jest rozmiar obiektu typu `int`?”.
- Wynik `sizeof` ma typ `size_t`. Do jego wypisywania w `printf` używamy formatu **`%zu`**. Nie musimy jeszcze znać szczegółów `size_t`.
- Rozmiary podstawowych typów zależą od **implementacji C i platformy docelowej**. Typowe `sizeof(int) == 4` to obserwacja na wielu współczesnych komputerach, nie gwarancja języka.
- W tym przykładzie compiler zna rozmiar typu już podczas kompilacji. Program **wyświetla rozmiar właściwy dla swojej platformy**, a nie odczytuje go dopiero z pamięci podczas działania.

**B** oznacza tutaj *bajt w rozumieniu C*. Czy bajt zawsze musi składać się z ośmiu bitów? Sprawdźmy.

### Eksperyment 2 — porównujemy rozmiary

**Przewidź**, zanim uruchomisz program:

1. Czy `sizeof(char)` da 1?
2. Czy `sizeof(int)`, `sizeof(float)` i `sizeof(double)` muszą być równe?
3. Ile **bitów** ma jeden bajt na naszej platformie?

`CHAR_BIT` z `<limits.h>` podaje liczbę bitów w bajcie C. Pozostałe rozmiary sprawdzimy przez `sizeof`.

Kod zawiera również `int liczba = 42;`, które poznaliśmy już w W1. Zobacz, co zwraca `sizeof liczba`.

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    int liczba = 42;

    printf("CHAR_BIT = %d bits per byte\n", CHAR_BIT);
    printf("sizeof(char) = %zu B\n", sizeof(char));
    printf("sizeof(int) = %zu B\n", sizeof(int));
    printf("sizeof(long long int) = %zu B\n", sizeof(long long int));
    printf("sizeof(float) = %zu B\n", sizeof(float));
    printf("sizeof(double) = %zu B\n", sizeof(double));
    printf("sizeof liczba = %zu B\n", sizeof liczba);

    return 0;
}
````

### Co wynika z porównania?

- W C **`sizeof(char)` zawsze wynosi 1** — taki jest sposób zdefiniowania jednostki „bajt”. Natomiast liczba bitów w jednym bajcie wynosi `CHAR_BIT` (na typowym współczesnym komputerze: 8).
- Pozostałych rozmiarów nie wpisujemy na stałe jako praw uniwersalnych. Wyniki zależą od implementacji.
- `sizeof(int)` pyta o **typ** (stąd nawiasy), zaś `sizeof liczba` pyta o rozmiar obiektu/wyrażenia. Nawiasy można także napisać jako `sizeof(liczba)`, ale wtedy nie są one wywołaniem funkcji.
- Jeżeli obiekt ma `n` bajtów, odpowiada mu `n * CHAR_BIT` bitów pamięci przeznaczonej na reprezentację. Sam rozmiar **nie mówi jeszcze wszystkiego** o zakresie wartości.

**Zmień tylko jedną rzecz**: `int liczba = 42;` na `double liczba = 42.0;`. Uruchom ponownie. Czy zmienia się `sizeof liczba`? Czy zmienia się też `sizeof(int)`?

Do sposobu reprezentowania `float` i `double` wrócimy dalej; teraz interesuje nas wyłącznie rozmiar.

### Co zabieramy z tego eksperymentu?

**Typ to nie dekoracja nazwy.** Wybór typu ma konsekwencje dla tego, jak compiler interpretuje wartości, ile miejsca potrzebują obiekty oraz jakie działania możemy wykonać.

Umiemy już **zmierzyć rozmiar zamiast wierzyć w tabelę**. Następny krok: utworzymy zmienne typu `int`, przypiszemy im wartości i sprawdzimy, jak ich typ wpływa na zapis i wynik obliczeń. Zakresy wartości i sytuacje błędne zbadamy w osobnych eksperymentach.

## `int` — deklaracja, inicjalizacja, przypisanie

Zmierzyliśmy już rozmiar obiektu typu `int`. Teraz zaczniemy **przechowywać i zmieniać wartości**. W W1 robiliśmy to między innymi w instrukcji `int wynik = sum(5, 6);`.

```c
int a = 90;  // deklaracja i inicjalizacja
int b;       // sama deklaracja
b = 10;      // przypisanie wartości
```

- **Deklaracja** `int b;` wprowadza nazwę i typ zmiennej.
- **Inicjalizacja** `int a = 90;` nadaje wartość przy deklaracji.
- **Przypisanie** `b = 10;` ustawia lub zmienia wartość istniejącej zmiennej; `=` nie oznacza tu sprawdzania równości.

**Pytanie:** jakie wartości wypisze program poniżej, zanim i po tym, jak zmienimy `a`? Zauważ, że `b` otrzymuje wartość **przed pierwszym odczytem**.

````c
#include <stdio.h>

int main(void)
{
    int a = 90;
    int b;
    b = 10;

    printf("a=%d, b=%d\n", a, b);
    a = 91;
    printf("after assignment: a=%d, b=%d\n", a, b);

    return 0;
}
````

### Obserwacja — typ zostaje, wartość się zmienia

`a` i `b` mają ten sam typ `int`, ale różne wartości. Instrukcja `a = 91;` zmienia **wartość zmiennej**, nie jej typ. W `printf` używamy `%d` do wypisania wartości typu `int` jako liczby dziesiętnej.

**Eksperyment:** zamień tylko `b = 10;` na `b = -10;`. Uruchom ponownie: czy zmiana znaku wymaga innego typu? Następnie przywróć `b = 10;`.

Na razie dbamy, żeby każda odczytywana zmienna miała już przypisaną wartość. Pytanie, co dzieje się **bez inicjalizacji**, zostawiamy do następnej części.

## Jedna wartość, różne zapisy w kodzie

**Pytanie:** czy zapisy `16`, `020` i `0x10` oznaczają trzy różne liczby?

W C liczby całkowite można zapisywać m.in. w systemie dziesiętnym (bez przedrostka), **ósemkowym** (początkowe `0`) i **szesnastkowym** (przedrostek `0x` lub `0X`). Tak zapisane fragmenty kodu to *literały całkowite*.

**Przewidź wynik:** jaką liczbę wypisze `printf("%d", ...)` dla każdej z trzech zmiennych?

````c
#include <stdio.h>

int main(void)
{
    int dziesietnie = 16;
    int osemkowo = 020;
    int szesnastkowo = 0x10;

    printf("%d %d %d\n", dziesietnie, osemkowo, szesnastkowo);
    return 0;
}
````

### Skąd wzięły się trzy szesnastki?

`16`, `020` i `0x10` są **różnymi zapisami tej samej wartości**. Przedrostek zmienia **sposób odczytania cyfr w kodzie źródłowym**, nie typ zmiennej ani samą liczbę.

`%d` nie „konwertuje zmiennej na liczbę dziesiętną” — sprawia, że `printf` wypisuje jej wartość **w zapisie dziesiętnym**.

**Eksperyment:** zmień tylko `0x10` na `0x11`, uruchom ponownie i wyjaśnij, dlaczego zmieniła się ostatnia wypisana liczba. Potem przywróć `0x10`.

````c
#include <stdio.h>

int main(void)
{
    int a = 20;
    int b = 020;
    int c = 0x20;

    printf("20 = %d\n", a);
    printf("020 = %d\n", b);
    printf("0x20 = %d\n", c);
    return 0;
}
````

### Te same cyfry, inne wartości

W drugim eksperymencie cyfry `20` występują w trzech systemach pozycyjnych:

- `20` to dwadzieścia (system dziesiętny),
- `020` to szesnaście (system ósemkowy),
- `0x20` to trzydzieści dwa (system szesnastkowy).

**Pytanie:** dlaczego `020` **nie** oznacza po prostu „20 z zerem na początku”?

Zapamiętaj regułę czytania literałów, a nie konkretne wyniki z tabeli.

## Ten sam obiekt, różny sposób wyświetlenia: `printf`

Znany z W1 zapis `printf("%d", a)` służy do wypisania `int` dziesiętnie. Dla **`unsigned int`**, czyli odmiany typu całkowitego bez wartości ujemnych, możemy zastosować różne specyfikatory:

- `%u` — zapis dziesiętny,
- `%o` — zapis ósemkowy,
- `%x` / `%X` — zapis szesnastkowy z małymi / wielkimi literami.

**Pytanie:** czy różne specyfikatory zmieniają wartość zmiennej `liczba`, czy tylko tekst wypisany na ekranie?

Na szczegóły zakresu `unsigned int` przyjdzie czas w kolejnej części. Na razie istotne jest, aby **typ argumentu pasował do specyfikatora** w `printf`.

````c
#include <stdio.h>

int main(void)
{
    unsigned int liczba = 16;

    printf("decimal: %u\n", liczba);
    printf("octal:   %o\n", liczba);
    printf("hex:     %x\n", liczba);
    printf("HEX:     %X\n", liczba);

    return 0;
}
````

### Wartość ≠ literał ≠ tekst wypisany przez `printf`

Porównajmy trzy rzeczy:

1. **Literał** — jak zapisaliśmy liczbę w kodzie źródłowym: `16`, `020` albo `0x10`.
2. **Wartość** — w tym przypadku zawsze ta sama, szesnaście.
3. **Format wypisania** — `%u`, `%o`, `%x` lub `%X` decyduje, jakie cyfry zobaczymy.

**Eksperyment:** zmień jedynie `unsigned int liczba = 16;` na `unsigned int liczba = 31;`. Przewidź zapisy dziesiętny, ósemkowy i szesnastkowy, a potem je porównaj.

**Uwaga:** `%d` oczekuje argumentu typu `int`, a `%u`, `%o` i `%x`/`%X` argumentu typu `unsigned int`. Nie mieszamy tych wymagań przypadkowo. W tym miejscu pokazujemy zaledwie dwa warianty typu całkowitego — pełniejsze omówienie zakresów będzie później.

### Co już umiemy?

- Zadeklarować `int`, nadać mu wartość przy deklaracji lub później przez przypisanie.
- Zmienić wartość bez zmiany typu.
- Rozpoznać zapis dziesiętny, ósemkowy i szesnastkowy literału.
- Odróżnić **wartość w programie** od jej **sposobu wyświetlania**.

**Pytanie:** jeżeli `a = 0x10` i `b = 020`, czy `a` i `b` mają tę samą wartość? Czy `printf` może wypisać obie w zapisie dziesiętnym?

Teraz zobaczymy, dlaczego poprawna inicjalizacja i ograniczony zakres typu mają znaczenie. To będzie przedmiotem W2.4.

## W2.4 — Ile mieści się w `int`?

### 1. Zmienna musi mieć wartość, zanim ją odczytamy

Wiemy już, że `int a = 90;` oznacza **deklarację i inicjalizację**, a `int b;` — tylko deklarację.

**Pytanie:** co wypisze ten fragment?

```c
int a = 90;
int b, c;
printf("a=%d, b=%d, c=%d\n", a, b, c); // b i c nie mają nadanych wartości!
```

To **błędny fragment**, nie wzór do uruchamiania: odczyt niezainicjalizowanych lokalnych `int` może prowadzić do *undefined behavior* (UB). Nawet jeśli zobaczymy zera albo inne liczby, nie możemy im ufać.

Zobaczmy poprawioną wersję.

````c
#include <stdio.h>

int main(void)
{
    int a = 90;
    int b, c;

    b = 10;
    c = 40;
    printf("a=%d, b=%d, c=%d\n", a, b, c);

    return 0;
}
````

**Obserwacja:** `b` i `c` są odczytywane **dopiero po przypisaniu wartości**. Otrzymujemy `a=90, b=10, c=40`.

**Eksperyment:** zamień `int b, c;` oraz dwa przypisania na jedną instrukcję `int b = 10, c = 40;`. Czy wynik się zmieni?

Wróć do oryginalnego kodu przed następnym przykładem.

### 2. Jakie wartości może przechowywać `int`?

W W2.2 sprawdziliśmy, ile bajtów zajmuje `int`. Teraz zapytamy o jego **zakres**.

Nagłówek `<limits.h>` podaje granice dla bieżącej implementacji C:

- `INT_MIN` — najmniejsza wartość `int`,
- `INT_MAX` — największa wartość `int`,
- `UINT_MAX` — największa wartość `unsigned int`.

**Pytanie:** jakie liczby otrzymamy na komputerze, na którym właśnie pracujemy?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    printf("INT_MIN = %d\n", INT_MIN);
    printf("INT_MAX = %d\n", INT_MAX);
    printf("UINT_MAX = %u\n", UINT_MAX);

    return 0;
}
````

W konfiguracji odniesienia odczytaliśmy `INT_MIN = -2147483648`, `INT_MAX = 2147483647` oraz `UINT_MAX = 4294967295`. To **wyniki dla tej platformy**, nie liczby gwarantowane dla każdej implementacji C.

Wiemy też z W2.2, że w konfiguracji odniesienia `sizeof(int) = 4` i `CHAR_BIT = 8`, więc obiekt `int` zajmuje **32 bity**.

**Pytanie:** skąd właśnie taki zakres i dlaczego `INT_MIN` ma wartość o jeden mniejszą co do modułu niż `INT_MAX`? Zajrzyjmy do reprezentacji bitowej.

### Jak zapisać liczbę w bitach? Model 8-bitowy

Każdy **bit** może przyjąć wartość `0` albo `1`. Osiem bitów daje `2^8 = 256` różnych wzorców.

Dla wygody najpierw przyjrzymy się **8-bitowemu modelowi liczb ze znakiem w kodzie uzupełnień do dwóch** (*two's complement*). To model do nauki — **nie** twierdzimy, że typ `int` w C zajmuje 8 bitów.

| Zapis binarny (8 bitów) | Wartość |
|:------------------------|--------:|
| `0000 0000` | 0 |
| `0000 0001` | 1 |
| `0111 1111` | 127 |
| `1000 0000` | −128 |
| `1000 0001` | −127 |
| `1111 1110` | −2 |
| `1111 1111` | −1 |

W takiej reprezentacji **najstarszy bit** (po lewej stronie zapisu binarnego) wskazuje, czy liczba jest nieujemna (`0`), czy ujemna (`1`). Nie jest jednak osobną „lampką znaku”: **uczestniczy w kodowaniu całej wartości**. Dla 8 bitów ma wagę `−128`.

**Pytanie:** dlaczego największa dodatnia wartość to `127`, a najmniejsza ujemna to `−128`?

**Odpowiedź:** W 8-bitowym kodzie uzupełnień do dwóch najstarszy bit ma wagę **−128**, a pozostałe siedem bitów ma dodatnie wagi **64, 32, 16, 8, 4, 2 i 1**.

- **Największa dodatnia wartość:** `0111 1111` = 64 + 32 + 16 + 8 + 4 + 2 + 1 = **127**. Najstarszy bit jest zerem, więc nie wnosi wartości −128.
- **Najmniejsza ujemna wartość:** `1000 0000` = −128 + 0 = **−128**. Wszystkie pozostałe bity są zerami.
- Pozostałe przykłady z tabeli: `1000 0001` = −128 + 1 = **−127** oraz `1111 1110` = −128 + 126 = **−2**.

Mamy zatem **128 wzorców nieujemnych** (od 0 do 127) i **128 ujemnych** (od −128 do −1). Zero należy do pierwszej grupy — stąd jeden dodatkowy wynik ujemny i brak dodatniego `+128` w tym 8-bitowym modelu.

### Jak wygląda nasz 32-bitowy `int`?

W konfiguracji odniesienia obiekt `int` zajmuje **4 bajty po 8 bitów**. Gdy jest przechowywany w pamięci, mieści się więc w nim 32-bitowy wzorzec. GCC na tej platformie używa dla `int` **uzupełnień do dwóch**.

| Wartość typu `int` | Wzorzec 32 bitów |
|:---------------------|:-----------------|
| `0` | `00000000 00000000 00000000 00000000` |
| `INT_MAX = 2147483647` | `01111111 11111111 11111111 11111111` |
| `INT_MIN = -2147483648` | `10000000 00000000 00000000 00000000` |
| `-1` | `11111111 11111111 11111111 11111111` |

**Zwróć uwagę na pierwszy bit od lewej**: dla `INT_MAX` jest równy `0`, a dla `INT_MIN` — `1`. W 32-bitowym kodzie uzupełnień do dwóch jego waga wynosi `−2^31`.

Odstępy co osiem bitów są tu **wyłącznie dla czytelności**. Tabela przedstawia **logiczny zapis liczby**, a nie kolejność adresów bajtów w pamięci. Kolejnością bajtów (*endianness*) zajmiemy się później.

### Te same bity — dlaczego typ ma znaczenie?

Na naszej 32-bitowej platformie wzorzec:

```
10000000 00000000 00000000 00000000
```

odczytujemy jako **`−2147483648`**, gdy interpretujemy go jako `int`, ale jako **`2147483648`**, gdy interpretujemy go jako `unsigned int`.

Podobnie wzorzec złożony z 32 jedynek odpowiada **`−1`** jako `int`, a **`4294967295`** jako `unsigned int`.

To pokazuje różnicę między **wzorcem bitów** a **wartością wynikającą z typu**. Nie należy z tego wnioskować, że każde przepełnienie `int` musi dawać określony wzorzec: standard C nie gwarantuje takiego wyniku.

**Pytanie:** jeśli zaczynamy od `INT_MAX`, którego zapis binarny to `0` i 31 jedynek, jaki 32-bitowy wzorzec mogłoby dać sprzętowe dodanie jedynki? Teraz sprawdzimy, co rzeczywiście wypisze program w konfiguracji odniesienia.

### 3. Przepełnienie `int` — sprawdzamy eksperymentalnie

To **celowy eksperyment z niepoprawną operacją**. W standardowym C, gdy suma liczb typu `int` przekracza dopuszczalny zakres, następuje **undefined behavior**.

Nie znamy więc wyniku zagwarantowanego przez język. Ale nadal możemy **uruchomić program i zobaczyć, co zrobi ta konkretna kompilacja**.

**Pytanie:** czy po dodaniu `1` zobaczymy liczbę ujemną, błąd, czy może coś innego?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    int a = INT_MAX;

    printf("przed=%d\n", a);
    a = a + 1;  // EKSPERYMENT: signed overflow to undefined behavior w C!
    printf("po=%d\n", a);

    return 0;
}
````

### Co zaobserwowaliśmy, a co mówi standard?

W konfiguracji odniesienia uruchomiliśmy program z przepełnieniem `int` i otrzymaliśmy:

```
przed=2147483647
po=-2147483648
```

Widzieliśmy przed chwilą, że dla naszej **32-bitowej reprezentacji uzupełnień do dwóch** odpowiada to zmianie wzorca:

```
  01111111 11111111 11111111 11111111   (INT_MAX)
+ 00000000 00000000 00000000 00000001   (+1)
---------------------------------------
  10000000 00000000 00000000 00000000   (INT_MIN)
```

Taki wynik można zrozumieć, patrząc na działanie 32-bitowego dodawania maszynowego i interpretację otrzymanych bitów. **To wyjaśnia obserwację, ale nie jest regułą standardu C.** Dla przepełnienia `int` standard mówi: **undefined behavior**.

Wynik zależy od kodu wygenerowanego przez kompilator i od warunków uruchomienia. Przy innych ustawieniach optymalizacji może być inny — choć nie musi. Sam wynik nie pozwala wskazać konkretnej optymalizacji.

**Wniosek:** najpierw widzimy, **co wykonał nasz program**; osobno ustalamy, **co gwarantuje język C**.

### 4. A co z `unsigned int`?

`unsigned int` nie przechowuje wartości ujemnych. Tutaj C **gwarantuje** arytmetykę modulo `UINT_MAX + 1`.

**Wyjaśnienie `1u`:** sufiks `u` oznacza literał typu `unsigned int` (w przeciwieństwie do `1`, które ma typ `int`). W tym przykładzie również `a + 1` da ten sam rezultat, ponieważ przy dodawaniu do `unsigned int a` jedynka zostanie przekonwertowana na `unsigned int`. Do zasad konwersji wrócimy w W2.7.

**Pytanie:** co wypisze bardzo podobny program, jeśli zaczniemy od `UINT_MAX` i dodamy `1u`?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    unsigned int a = UINT_MAX;

    printf("przed=%u\n", a);
    a = a + 1u;
    printf("po=%u\n", a);

    return 0;
}
````

**Obserwacja:** po przekroczeniu `UINT_MAX` otrzymujemy `0`. Tym razem **nie jest to przypadek UB** — takie zachowanie gwarantuje C.

Porównaj dwa eksperymenty:

- `int`: widzieliśmy określony wynik na konkretnej maszynie, ale **język go nie gwarantuje**;
- `unsigned int`: również zobaczyliśmy zawinięcie, ale **język gwarantuje** wynik modulo zakresu.

**Eksperyment:** zmień `unsigned int a = UINT_MAX;` na `unsigned int a = UINT_MAX - 1u;`. Co powinno się wypisać po dodaniu `1u`?

### Co wynika z W2.4?

- Wartość zmiennej lokalnej należy ustalić **zanim ją odczytamy**.
- Zakresy `int` i `unsigned int` można sprawdzić przez `<limits.h>`.
- Przepełnienie `int` to **undefined behavior** — eksperymentalny wynik nie jest gwarancją standardu.
- Dla `unsigned int` wynik zawija się modulo zakresu **zgodnie ze standardem**.

**Pytanie:** skoro `int` przechowuje liczby całkowite, to co będzie wynikiem `5 / 6`? W W2.5 przyjrzymy się działaniom arytmetycznym na takich liczbach.

## W2.5 — Działania na liczbach całkowitych

### Jeszcze o bitach: reprezentacja `unsigned int`

W W2.4 oglądaliśmy **8-bitowy model liczb ze znakiem** w kodzie uzupełnień do dwóch. Jak te same bity odczytamy jako **liczbę bez znaku**?

Dla **modelu 8-bitowego bez znaku** wszystkie bity mają dodatnie wagi: **128, 64, 32, 16, 8, 4, 2, 1**.

| Zapis binarny (8 bitów) | Wartość bez znaku |
|:-----------------------|------------------:|
| `0000 0000` | 0 |
| `0000 0001` | 1 |
| `0111 1111` | 127 |
| `1000 0000` | 128 |
| `1111 1111` | 255 |

**Pytanie:** dlaczego `1000 0000` w modelu ze znakiem oznaczało **−128**, a tutaj oznacza **128**? Bity się nie zmieniły — zmieniła się ich **interpretacja**.

W konfiguracji odniesienia `unsigned int` zajmuje 32 bity, zatem jego zakres to **od 0 do `2^32 − 1 = 4294967295`**, a najstarszy bit ma wagę **`2^31`**. Wzorzec 32 jedynek oznacza tutaj `UINT_MAX`. To konkretny przypadek implementacji w konfiguracji odniesienia, nie wymóg, aby wszędzie `unsigned int` miał 32 bity.

W W2.4 zauważyliśmy, że po `UINT_MAX + 1u` otrzymujemy 0. W C dla liczb bez znaku jest to **gwarantowane** działanie modulo liczby wszystkich możliwych wartości.

### Dodawanie, odejmowanie i mnożenie

Dotychczas przechowywaliśmy i wypisywaliśmy liczby. Teraz wykonamy działania:

- `+` — dodawanie;
- `-` — odejmowanie;
- `*` — mnożenie.

**Pytanie:** jeśli `a = 5` i `b = 4`, jakie wyniki wypisze program?

````c
#include <stdio.h>
int main(void)
{
    int a = 5, b = 4;

    printf("a=%d, b=%d\n", a, b);
    printf("a+b=%d\n", a + b);
    printf("a-b=%d\n", a - b);
    printf("a*b=%d\n", a * b);

    return 0;
}
````

**Obserwacja:** otrzymujemy `9`, `1` i `20`. Wszystkie działania wykonujemy na `int` i wypisujemy wyniki za pomocą `%d`.

**Eksperyment:** zmień tylko `b = 4` na `b = 6`. Przewidź wyniki i porównaj z programem.

Pamiętajmy o W2.4: przy arytmetyce `int` matematyczny wynik nie może przekroczyć dopuszczalnego zakresu, jeżeli oczekujemy zachowania zdefiniowanego przez C.

### Dzielenie `/` i reszta z dzielenia `%`

**Pytanie:** jaki wynik da `5 / 6`, jeśli obie liczby są typu `int`?

W C dzielenie **dwóch liczb całkowitych** daje wynik całkowity: część ułamkowa jest **odrzucana w kierunku zera**. Zatem `5 / 6` to `0`, a `6 / 5` to `1`.

Operator `%` pozwala obliczyć **resztę z dzielenia**, np. `6 % 5` daje `1`. Aby wydrukować dosłowny znak procenta wewnątrz napisu `printf`, piszemy `%%`.

Przewidź cztery wyniki, zanim uruchomisz kod.

````c
#include <stdio.h>
int main(void)
{
    int a = 5, b = 6;

    printf("a/b=%d\n", a / b);
    printf("b/a=%d\n", b / a);
    printf("a%%b=%d\n", a % b);
    printf("b%%a=%d\n", b % a);

    return 0;
}
````

**Obserwacja:** `5 / 6 = 0`, `6 / 5 = 1`, `5 % 6 = 5` i `6 % 5 = 1`.

Dzielenie całkowite **nie zaokrągla do najbliższej liczby**. Dla dodatnich argumentów po prostu odrzuca część ułamkową.

**Eksperyment:** zmień `a = 5` na `a = 13`. Co otrzymasz dla `a / b` i `a % b`?

**Uwaga:** dzielenie przez `0` i obliczanie reszty z dzielenia przez `0` są w C błędnymi operacjami (*undefined behavior*). Nie eksperymentujemy z nimi tutaj.

### A co z liczbami ujemnymi?

W C dzielenie całkowite odrzuca część ułamkową **w kierunku zera**, także dla liczb ujemnych. Dlatego `−7 / 3` daje `−2`, a nie `−3`.

Reszta spełnia zależność `a = (a / b) * b + (a % b)`, gdy dzielenie jest zdefiniowane. Jeśli reszta nie jest zerem, ma **znak dzielnej** (pierwszej liczby).

**Pytanie:** jakie wyniki otrzymamy dla `−7 % 3` i `7 % −3`?

````c
#include <stdio.h>
int main(void)
{
    int a = -7, b = 3;

    printf("%d/%d=%d\n", a, b, a / b);
    printf("%d%%%d=%d\n", a, b, a % b);
    printf("%d/%d=%d\n", -a, -b, (-a) / (-b));
    printf("%d%%%d=%d\n", -a, -b, (-a) % (-b));

    return 0;
}
````

**Obserwacja:** `−7 / 3 = −2`, `−7 % 3 = −1`, `7 / −3 = −2` i `7 % −3 = 1`.

Sprawdźmy: `−7 = (−2) * 3 + (−1)`. Wartość `−1` jest właśnie resztą z dzielenia `−7` przez `3`.

**Pytanie:** skoro `5 / 6` wykonane na `int` daje `0`, jak uzyskać wynik z częścią ułamkową?

Do tego potrzebujemy typów **`float`** i **`double`**. W W2.6 pokażemy nie tylko ich działanie, ale również **reprezentację bitową** (znak, wykładnik i część ułamkową; szczegóły standardu poznamy za chwilę) oraz związane z nią ograniczenia dokładności.

## W2.6 — `float` i `double`: liczby z częścią ułamkową

W W2.5 stwierdziliśmy, że `5 / 6` daje `0`, gdy oba argumenty są typu `int`. Jak otrzymać przybliżoną wartość **0,8333…**?

Do zapisywania liczb z częścią ułamkową używamy m.in. typów **`float`** i **`double`**.

- `5.0` jest w C literałem typu **`double`**.
- `5.0f` jest literałem typu **`float`**; sufiks `f` wskazuje typ, tak jak poznane wcześniej `u` w `1u`.

**Pytanie:** co wypisze ten sam iloraz obliczony na trzech typach?

````c
#include <stdio.h>

int main(void)
{
    int ai = 5, bi = 6;
    float af = 5.0f, bf = 6.0f;
    double ad = 5.0, bd = 6.0;

    printf("int:    %d\n", ai / bi);
    printf("float:  %.6f\n", af / bf);
    printf("double: %.6f\n", ad / bd);

    return 0;
}
````

**Obserwacja:** `int` daje `0`, natomiast `float` i `double` wypisują `0.833333` — z dokładnością do sześciu miejsc po przecinku.

**Ważne:** taki sam tekst wypisany na ekranie nie oznacza jeszcze, że obie wartości zmiennoprzecinkowe przechowano identycznie.

W `printf` format **`%.6f`** oznacza sześć cyfr po kropce; `%e` wykorzystamy później do zapisu naukowego. Do `printf` możemy stosować **`%f` zarówno dla `float`, jak i `double`**: argument typu `float` jest w takim wywołaniu automatycznie przekazywany jako `double`. `%lf` w `printf` (od C99) oznacza to samo co `%f`; nie należy tej reguły przenosić na `scanf`.

**Eksperyment:** zmień w pierwszej parze `ai = 5` na `ai = 12`. Czy pozostałe dwa wyniki się zmienią?

### Ile pamięci zajmują `float` i `double`?

W W2.2 poznaliśmy `sizeof`. Teraz sprawdźmy, **ile bitów dokładności** otrzymujemy w typach zmiennoprzecinkowych.

Nagłówek `<float.h>` udostępnia m.in. `FLT_RADIX`, `FLT_MANT_DIG`, `DBL_MANT_DIG`, `FLT_DIG` i `DBL_DIG`. Nie musimy jeszcze znać szczegółów tych nazw — odczytajmy wyniki.

**Pytanie:** czy dwa razy większa liczba bajtów oznacza po prostu „dwa razy więcej cyfr po przecinku”?

````c
#include <stdio.h>
#include <float.h>

int main(void)
{
    printf("sizeof(float)  = %zu B\n", sizeof(float));
    printf("sizeof(double) = %zu B\n", sizeof(double));
    printf("FLT_RADIX = %d\n", FLT_RADIX);
    printf("FLT_MANT_DIG = %d\n", FLT_MANT_DIG);
    printf("DBL_MANT_DIG = %d\n", DBL_MANT_DIG);
    printf("FLT_DIG = %d\n", FLT_DIG);
    printf("DBL_DIG = %d\n", DBL_DIG);

    return 0;
}
````

### Jak zapisujemy wartości zmiennoprzecinkowe w bitach?

Na typowej współczesnej platformie, również w konfiguracji odniesienia, `float` i `double` korzystają z formatów opisanych przez **IEEE 754** (*Institute of Electrical and Electronics Engineers — Instytut Inżynierów Elektryków i Elektroników*). **754 to numer standardu** poświęconego arytmetyce zmiennoprzecinkowej (*Standard for Floating-Point Arithmetic*), a nie liczba bitów. Dwa popularne formaty tego standardu to **binary32** (`float`) i **binary64** (`double`).

| Pole reprezentacji | `float` (binary32) | `double` (binary64) |
|:-------------------|--------------------:|---------------------:|
| **Znak** | 1 bit | 1 bit |
| **Wykładnik** | 8 bitów | 11 bitów |
| **Część ułamkowa** | 23 bity | 52 bity |
| **Łącznie** | **32 bity** | **64 bity** |

**`float`**: `[znak: 1] [wykładnik: 8] [część ułamkowa: 23]`  
**`double`**: `[znak: 1] [wykładnik: 11] [część ułamkowa: 52]`

Liczba nie jest tu przechowywana tak jak zwykły `int`. Dla zwykłych, niezerowych liczb **znormalizowanych** jej wartość można opisać wzorem: **`(−1)^S × (1.F)₂ × 2^(E − bias)`**. `S` oznacza bit znaku, `E` — liczbę zapisaną w polu wykładnika, `F` — bity części ułamkowej, a `bias` — stałe przesunięcie wykładnika. Symbol **`(1.F)₂` jest umownym zapisem binarnym**: przed kropką stoi **domyślna jedynka**, a `F` oznacza bity po kropce, nie liczbę dziesiętną. Nie zapisujemy tej jedynki w pamięci. Na przykład gdy `F` zaczyna się od `101`, a dalej są same zera, otrzymujemy **`(1.101)₂ = 1 + 1/2 + 1/8 = 1.625`**. Dziesiętną wartość takiej znaczącej oznaczymy literą **`m`**.

To przypomina **notację naukową w systemie dwójkowym**, ale zamiast potęg dziesiątki używamy potęg **dwójki**. Wykładnik i zakres jego wartości wyjaśnimy na następnych przykładach.

Standard C **nie wymaga**, aby każda implementacja używała dokładnie tych formatów. W dalszych przykładach wyjaśniamy reprezentację właściwą dla naszej platformy. W następnym bloku zobaczymy ważny wyjątek: dla **zera i liczb podnormalnych** ukryta jedynka nie występuje.

### Wyjątek: skąd bierze się zero, skoro liczby mają ukrytą jedynkę?

Wzór `(−1)^S × (1.F)₂ × 2^(E − bias)` dotyczy **wyłącznie liczb znormalizowanych**. Gdy **`E = 0`**, **nie ma ukrytej jedynki**.

| Znak S | Wykładnik E | Część ułamkowa F | Znaczenie |
|:------:|:------------|:-----------------|:----------|
| `0` | same zera | same zera | **`+0.0`** |
| `1` | same zera | same zera | **`−0.0`** |
| `0` lub `1` | same zera | **nie same zera** | **liczba podnormalna** |

**Reguła interpretacji wykładnika nie jest zawsze taka sama:**

- Gdy `1 ≤ E ≤ 2^k − 2` (`k` to liczba bitów pola wykładnika), liczba jest **znormalizowana**, ma ukrytą jedynkę i wykładnik **`e = E − bias`**.
- Gdy **`E = 0` i `F ≠ 0`**, liczba jest **podnormalna**: **nadal nie ma ukrytej jedynki**, a znacząca to **`m = (0.F)₂`**, z wykładnikiem **`e = 1 − bias`**, a nie `−bias`.
- Gdy **`E = 0` i `F = 0`**, mamy zero ze znakiem wyznaczanym przez `S`.
- Gdy `E` składa się z samych jedynek, jest to zapis **nieskończoności lub NaN**, a nie zwykłej liczby z wykładnikiem `E − bias`.

**Przykład dla `float`:** `E = 00000000`, `F = 000…001` (jedynka tylko na ostatnim z 23 miejsc) daje `m = 2^(−23)` i `e = 1 − 127 = −126`. Otrzymujemy najmniejszą dodatnią liczbę podnormalną **`2^(−23) × 2^(−126) = 2^(−149)`**. Przy `F = 0` dostalibyśmy **zero**, nie liczbę z ukrytą jedynką.

**Pytanie:** czy samo `E = 0` wystarczy, żeby zapisać zero? **Nie.** Trzeba również ustawić **`F = 0`**. O znaku zera decyduje bit `S`.

W kolejnych tabelach kolumna **„Znacząca m (dziesiętnie)”** będzie oznaczała dodatnią wartość `m = (1.F)₂ = 1 + F/2^p` dla liczb znormalizowanych, gdzie `p = 23` dla `float` i `p = 52` dla `double`. Dla zera wpisujemy **0 (wyjątek)**. To wartość **znaczącej**, nie wynikowa wartość całej liczby — trzeba jeszcze zastosować znak i wykładnik.

### Przykład: jak zakodować `1.0` i `0.5`?

Spójrzmy na dwie liczby, które **można dokładnie zapisać binarnie**:

```
1.0 = 1.0 (binarnie) × 2^0
0.5 = 1.0 (binarnie) × 2^(-1)
```

W polu wykładnika nie przechowujemy bezpośrednio potęgi `2`. Dla liczb **znormalizowanych** zapisujemy **`E = e + bias`**, gdzie `e` jest właściwym wykładnikiem, zaś *bias* to **przesunięcie**: 127 dla `float` i 1023 dla `double`.

**Odczyt:** `e = E − bias`. Zatem dla `float` **127 → `2^0`**, **126 → `2^(−1)`**, a dla `double` **1023 → `2^0`**, **1022 → `2^(−1)`**. Ta sama liczba `1023` nie jest wykładnikiem liczby `float`!

| Wartość | Znak S | Wykładnik E — `float` (8 bitów) | Wykładnik E — `double` (11 bitów) | Część ułamkowa F | Znacząca m (dziesiętnie) |
|:--------|:------:|:----------------------------------|:------------------------------------|:-----------------|------------------------:|
| `+0.0` | `0` | `00000000` (0) | `00000000000` (0) | same zera | **0** (wyjątek) |
| `−0.0` | `1` | `00000000` (0) | `00000000000` (0) | same zera | **0** (wyjątek) |
| `1.0` | `0` | `01111111` (127) | `01111111111` (1023) | same zera | **1.0** |
| `0.5` | `0` | `01111110` (126) | `01111111110` (1022) | same zera | **1.0** |
| `−1.0` | `1` | `01111111` (127) | `01111111111` (1023) | same zera | **1.0** |
| `−0.5` | `1` | `01111110` (126) | `01111111110` (1022) | same zera | **1.0** |

Dla **liczb znormalizowanych** w tej tabeli ukryta jedynka wraz z zerami w `F` daje znaczącą **`m = 1.0`**. Zmiana rzeczywistego wykładnika z 0 na −1 dzieli wartość przez dwa. **Dla zera jest inaczej**: `E = 0`, `F = 0` i **`m = 0`**, bez ukrytej jedynki.

**Pytanie:** jak zmienić `1.0` w `−1.0` bez zmiany jej wielkości? W tych formatach wystarczy zmienić **bit znaku** z `0` na `1`.

To logiczny zapis pól liczby, **nie** kolejność bajtów pod adresami pamięci.

### Minimalny i maksymalny wykładnik — jakie wartości mieszczą się w polu?

Skoro wykładnik zapisujemy na skończonej liczbie bitów, także jego zakres jest ograniczony. **Uwaga:** poniższe granice dotyczą **zwykłych liczb znormalizowanych** w formatach IEEE 754. Skrajne kombinacje bitów wykładnika mają szczególne znaczenie.

| Własność | `float` (binary32) | `double` (binary64) |
|:---------|--------------------:|---------------------:|
| Liczba bitów pola `E` | 8 | 11 |
| Wszystkie możliwe `E` | 0–255 | 0–2047 |
| Przesunięcie `bias` | 127 | 1023 |
| `E` dla liczb znormalizowanych | **1–254** | **1–2046** |
| Minimalny rzeczywisty wykładnik `e = E − bias` | **−126** | **−1022** |
| Maksymalny rzeczywisty wykładnik `e = E − bias` | **+127** | **+1023** |

**Przykład:** dla `float` pole `E = 1` daje **`2^(−126)`**, a `E = 254` daje **`2^127`**. Dla `double` analogiczne granice to **`2^(−1022)`** oraz **`2^1023`**. Pamiętajmy, że na całą wartość wpływa jeszcze część znacząca `(1.F)₂` — sama potęga nie jest maksimum liczby.

Co z pozostałymi wzorcami? **Same zera w polu `E`** oznaczają zero lub liczby **podnormalne** (bardzo bliskie zeru), które umożliwiają zapis wartości mniejszych od najmniejszej liczby znormalizowanej. **Same jedynki w polu `E`** oznaczają nieskończoność albo **NaN** (*Not a Number* — wartość „nie-liczba”), zależnie od bitów `F`. Dla tych przypadków **nie stosujemy wprost** wzoru `(1.F)₂ × 2^(E − bias)`.

**Pytanie:** dlaczego `double` może przechowywać zarówno liczby znacznie mniejsze, jak i znacznie większe od `float`?

### Liczby podnormalne — co dzieje się na granicy?

Znamy już najmniejszą dodatnią liczbę podnormalną `float`: `2^(−149)`. Porównajmy ją z **drugą najmniejszą**, następnie z liczbą **w środku zakresu podnormalnego**, a na końcu z **najmniejszą znormalizowaną**.

Dla wszystkich podnormalnych `E = 0`, **nie ma ukrytej jedynki**, a rzeczywisty wykładnik to `e = 1 − 127 = −126`. Zmieniamy tylko **23 bity pola `F`**; znacząca ma postać `m = (0.F)₂`. Dla pierwszej liczby znormalizowanej `E = 1` i **wraca ukryta jedynka**: `m = (1.F)₂`.

| Przypadek (`float`) | Wykładnik E (8 bitów) | Część ułamkowa F (23 bity) | Znacząca m | Wartość |
|:---------------------|:----------------------|:---------------------------|:-----------|:--------|
| **Najmniejsza dodatnia podnormalna** | `00000000` | `00000000000000000000001` | `2^(−23)` | `2^(−149) ≈ 1.401298464e−45` |
| **Druga najmniejsza dodatnia podnormalna** | `00000000` | `00000000000000000000010` | `2^(−22)` | `2^(−148) ≈ 2.802596929e−45` |
| **W środku** zakresu podnormalnego | `00000000` | `10000000000000000000000` | `0.5` | `2^(−127) ≈ 5.877471754e−39` |
| **Największa podnormalna** | `00000000` | `11111111111111111111111` | `1 − 2^(−23)` | `2^(−126) − 2^(−149) ≈ 1.175494211e−38` |
| **Najmniejsza znormalizowana** | `00000001` | `00000000000000000000000` | `1.0` (ukryta jedynka) | `2^(−126) ≈ 1.175494351e−38` |

**Pytanie:** czy pomiędzy największą liczbą podnormalną i najmniejszą znormalizowaną pojawia się duża przerwa?

**Nie.** Różnią się dokładnie o `2^(−149)` — tyle samo, ile najmniejsza dodatnia liczba podnormalna. Reprezentacja zachowuje więc ciągłość **kroków między sąsiednimi dodatnimi wartościami** przy tej granicy. Nie oznacza to, że można przedstawić dowolną liczbę rzeczywistą: reprezentowalnych wartości jest wciąż skończenie wiele.

Porównaj pierwsze dwa wiersze: różnica między ich wartościami wynosi dokładnie `2^(−149)`. Dla liczb podnormalnych odstępy między sąsiednimi dodatnimi wartościami są **stałe** (`2^(−149)` dla binary32), a **względna** precyzja maleje wraz ze zbliżaniem się do zera. Graniczną wartość znormalizowaną można odczytać z `FLT_MIN` w `<float.h>` — jest to najmniejsza **dodatnia znormalizowana** wartość `float`.

### Przykłady kodowania w `float`: liczby ujemne, π i `100.256`

Przećwiczmy podział na **znak `S` (1 bit)**, **wykładnik `E` (8 bitów)** i **część ułamkową `F` (23 bity)**.

| Wartość | Znak S | Wykładnik E (8 bitów) | Część ułamkowa F (23 bity) | Znacząca m (dziesiętnie) |
|:--------|:------:|:------------------------|:---------------------------|-------------------------:|
| `+0.0f` | `0` | `00000000` (0) | `00000000000000000000000` | **0** (wyjątek) |
| `−0.0f` | `1` | `00000000` (0) | `00000000000000000000000` | **0** (wyjątek) |
| `1.0f` | `0` | `01111111` (127) | `00000000000000000000000` | **1.0** |
| `−1.0f` | `1` | `01111111` (127) | `00000000000000000000000` | **1.0** |
| `π ≈ 3.1415927f` | `0` | `10000000` (128) | `10010010000111111011011` | ≈ **1.5707963705063** |
| `−π ≈ −3.1415927f` | `1` | `10000000` (128) | `10010010000111111011011` | ≈ **1.5707963705063** |
| `100.256f` | `0` | `10000101` (133) | `10010001000001100010010` | ≈ **1.5664999485016** |

**Skąd wykładnik?** Ponieważ `2 ≤ π < 4`, mamy `π ≈ 1.1001001…₂ × 2^1`, więc **`E = 1 + 127 = 128`**. Ponieważ `64 ≤ 100.256 < 128`, otrzymujemy **`e = 6`**, czyli **`E = 6 + 127 = 133`**.

**Pytanie:** co zmieni się między `π` i `−π`? **Tylko bit znaku** — pozostałe pola oraz dodatnia wartość znaczącej `m` są identyczne. Dla przykładu `π` w `float` daje `+1 × 1.5707963705063… × 2^1 ≈ 3.1415927410126`, a `−π` różni się tylko znakiem.

**Ważne:** zarówno π, jak i `100.256` są tu **zaokrąglane** do najbliższej wartości reprezentowalnej w `float`. Na naszej platformie `100.256f` jest przechowywane w przybliżeniu jako **`100.2559967041015625`**. Nie oznacza to, że `100.256` „zmieniło definicję” — zmieniło się jego przybliżenie w ograniczonej liczbie bitów.

### Te same przykłady w `double`: więcej bitów oznacza lepszą dokładność

W `double` mamy **1 bit znaku**, **11 bitów pola wykładnika** i **52 bity części ułamkowej**. Pole `F` jest długie, więc wyświetlamy je **szesnastkowo (hex)**. Jedna cyfra szesnastkowa oznacza **dokładnie cztery bity**, np. `0 → 0000`, `1 → 0001`, `2 → 0010`, `9 → 1001`, `b → 1011`, `d → 1101`, `f → 1111`.

**Przykład odczytu `F` dla π:** bierzemy 13 cyfr hex `921fb54442d18` i **każdą z nich zamieniamy na cztery bity**, zachowując kolejność:

`hex: 9 2 1 f b 5 4 4 4 2 d 1 8`  
`bity: 1001 0010 0001 1111 1011 0101 0100 0100 0100 0010 1101 0001 1000`

Otrzymujemy dokładnie **52 bity pola `F`**, nie całej liczby. Wartości `S` i `E` podajemy osobno. Ten zapis nie mówi nic o kolejności bajtów pod adresami pamięci.

| Wartość | Znak S | Wykładnik E (11 bitów) | Część ułamkowa F (52 bity; hex) | Znacząca m (dziesiętnie) |
|:--------|:------:|:-------------------------|:--------------------------------|-------------------------:|
| `+0.0` | `0` | `00000000000` (0) | `0000000000000` | **0** (wyjątek) |
| `−0.0` | `1` | `00000000000` (0) | `0000000000000` | **0** (wyjątek) |
| `1.0` | `0` | `01111111111` (1023) | `0000000000000` | **1.0** |
| `−1.0` | `1` | `01111111111` (1023) | `0000000000000` | **1.0** |
| `π ≈ 3.141592653589793` | `0` | `10000000000` (1024) | `921fb54442d18` | ≈ **1.5707963267948966** |
| `−π ≈ −3.141592653589793` | `1` | `10000000000` (1024) | `921fb54442d18` | ≈ **1.5707963267948966** |
| `100.256` | `0` | `10000000101` (1029) | `910624dd2f1aa` | ≈ **1.566500000000000004** |

Dla `π` rzeczywisty wykładnik to **`1024 − 1023 = 1`**, a dla `100.256` — **`1029 − 1023 = 6`**. Są takie same jak dla `float`. Różnica polega przede wszystkim na **znacznie dłuższym polu `F`** i większym możliwym zakresie wykładników.

W tabeli dziesiętna znacząca to `m = 1 + F/2^52`, przy czym pole `F` odczytujemy po zamianie hex na 52 bity. Dla π mamy `m ≈ 1.5707963267948966`; po pomnożeniu przez `2^1` otrzymujemy przybliżenie π. Dla `100.256` mamy `m ≈ 1.566500000000000004`; po pomnożeniu przez `2^6` otrzymujemy przechowywaną wartość.

Liczba `100.256` również **nie ma dokładnej reprezentacji w binary64**. Na naszej platformie zapisana wartość wynosi około **`100.25600000000000023`** — dużo bliżej wartości dziesiętnej niż dla `float`.

**Pytanie:** skoro `π` w obu typach ma ten sam wykładnik `e = 1`, dlaczego `double` podaje więcej cyfr π? Bo więcej bitów przeznaczono na **cyfry znaczące**, a nie na samą potęgę dwójki.

Pokazujemy tu **logiczne pola reprezentacji**, nie kolejność bajtów w pamięci (*endianness*).

### Czy każda liczba dziesiętna ma dokładny zapis binarny?

**Pytanie:** dlaczego `0.5` można zapisać dokładnie, ale z `0.1` jest trudniej?

`0.5 = 1/2`, więc w systemie dwójkowym ma skończony zapis: **`0.1₂`**. Indeks dolny **`₂` oznacza podstawę 2** (system dwójkowy), a **`₁₀` — podstawę 10** (system dziesiętny). Dlatego **`0.1₂ = 1 × 2^(−1) = 0.5₁₀`**. To NIE jest dziesiętne `0.1`!

Natomiast **`0.1₁₀`**, czyli jedna dziesiąta, ma w systemie dwójkowym **nieskończone rozwinięcie okresowe**:

```
0.1 (dziesiętnie) ≈ 0.00011001100110011... (binarnie)
```

Typy `float` i `double` mogą przechować tylko skończoną liczbę bitów, więc zapisują **przybliżenia** tej wartości.

Przewidź, czy oba typy wypiszą dokładnie `0.1`, gdy poprosimy o większą liczbę cyfr po przecinku.

````c
#include <stdio.h>

int main(void)
{
    float a = 0.1f;
    double b = 0.1;

    printf("float:  %.9f\n", a);
    printf("double: %.17f\n", b);

    return 0;
}
````

**Obserwacja w konfiguracji odniesienia:**

```
float:  0.100000001
double: 0.10000000000000001
```

Program wypisuje dziesiętne rozwinięcia **przybliżeń przechowywanych w pamięci**. To nie oznacza, że dodaliśmy do `0.1` jakieś wartości w kodzie.

`float` ma mniej bitów precyzji, więc jego przybliżenie jest tu bardziej oddalone od dokładnej liczby dziesiętnej `0.1` niż przybliżenie zapisane w `double`.

Format `%.9f` wypisuje **9 cyfr po kropce**, a `%.17f` — **17 cyfr po kropce**. Liczba miejsc po kropce nie jest tym samym co liczba poprawnych **cyfr znaczących**.

**Eksperyment:** zmień `0.1f` na `0.5f` oraz `0.1` na `0.5`. Czy teraz otrzymasz dokładne połówki?

### Dlaczego `double` ma większą precyzję?

W typowych formatach IEEE 754 liczby znormalizowane mają:

- `float`: **24 bity cyfr znaczących** (23 zapisane i 1 ukryty) — około **7 cyfr dziesiętnych**;
- `double`: **53 bity cyfr znaczących** (52 zapisane i 1 ukryty) — około **15–16 cyfr dziesiętnych**.

To **cyfry znaczące**, a nie stała liczba miejsc po przecinku! Wartość `FLT_DIG` i `DBL_DIG` z `<float.h>` pomaga określić gwarantowaną precyzję dziesiętną.

**Pytanie:** czy dodanie `1` do bardzo dużej liczby zawsze zmieni wartość typu `float`? Sprawdźmy na dokładnie reprezentowanej liczbie `2^24 = 16777216`.

````c
#include <stdio.h>

int main(void)
{
    float a = 16777216.0f;
    float a1 = a + 1.0f;
    double b = 16777216.0;
    double b1 = b + 1.0;

    printf("float:  %.0f -> %.0f\n", a, a1);
    printf("double: %.0f -> %.0f\n", b, b1);

    return 0;
}
````

**Obserwacja w konfiguracji odniesienia:**

```
float:  16777216 -> 16777216
double: 16777216 -> 16777217
```

To **nie jest przepełnienie zakresu**, lecz **utrata precyzji**: w tym rejonie wartości kolejne liczby reprezentowalne przez `float` są oddalone od siebie bardziej niż o 1. Po dodaniu jedynki i zaokrągleniu do `float` wartość pozostaje niezmieniona.

`double` ma więcej bitów znaczących i potrafi w tym przypadku odróżnić obie liczby.

**Eksperyment:** zmień `16777216.0f` na `100.0f` oraz `16777216.0` na `100.0`. Czy w obu typach dodanie jedynki zostanie teraz zauważone?

### Działania na `double` i notacja naukowa

Na liczbach zmiennoprzecinkowych możemy wykonywać znane operacje: `+`, `-`, `*` i `/`.

W `printf` można korzystać z formatu `%e`, czyli **notacji naukowej**. Podobnie w kodzie źródłowym zapis `6.02214076e23` oznacza liczbę około `6.02214076 × 10^23` — `e23` jest częścią **zapisu dziesiętnego** literału, a nie polem wykładnika IEEE 754 w pamięci.

**Pytanie:** czy `4.5 / 1.5` da wynik `3`, czy `3.0`? Jak pokaże go `printf("%.2f", ...)`?

````c
#include <stdio.h>

int main(void)
{
    double a = 4.5, b = 1.5;

    printf("a+b=%.2f\n", a + b);
    printf("a-b=%.2f\n", a - b);
    printf("a*b=%.2f\n", a * b);
    printf("a/b=%.2f\n", a / b);
    printf("scientific: %e\n", 6.02214076e23);

    return 0;
}
````

### Co zabieramy z W2.6?

- `5 / 6` dla `int` daje `0`, ale `5.0 / 6.0` daje wynik z częścią ułamkową.
- Typowe `float` i `double` korzystają z formatu złożonego z **bitu znaku, wykładnika i części ułamkowej**, a nie z zapisu całkowitego jak `int`.
- Nie każdą liczbę dziesiętną da się zapisać dokładnie w skończonej reprezentacji binarnej.
- `double` zwykle zachowuje więcej **cyfr znaczących** niż `float`.
- Zapis `0.1f` wybiera `float`, a `0.1` bez sufiksu jest literałem `double`.
- `%.6f` i `%e` zmieniają **sposób wyświetlenia**, a nie typ przechowywanej zmiennej.

**Pytanie:** co zrobi C, jeśli w jednym wyrażeniu spotkają się różne typy, na przykład `int` i `double`? W W2.7 zajmiemy się przypisaniem, konwersjami i utratą informacji.

## W2.7 — Konwersje typów i utrata informacji

W W2.5 wynik `1 / 2` był równy `0`. W W2.6 zobaczyliśmy, że `1.0 / 2.0` daje `0.5`. **Dlaczego typy wpływają na wynik działania?**

W C wartości mogą być przekształcane między typami:
- **automatycznie (niejawnie)**, np. gdy w jednym działaniu spotykają się `int` i `double` albo gdy przypisujemy `double` do `int`;
- **na nasze żądanie (jawnie)** przez zapis **`(typ)wyrażenie`**, np. `(double)a` lub `(int)x`. To właśnie *rzutowanie* (*cast*).

Nie każda konwersja zachowuje wszystkie informacje. Zaczniemy od przykładu, którego wynik potrafimy już przewidzieć.

**Pytanie:** czym różnią się `(double)a / b` oraz `(double)(a / b)`?

````c
#include <stdio.h>

int main(void)
{
    int a = 1, b = 2;

    printf("int: %d\n", a / b);
    printf("double literal: %.2f\n", 1.0 / 2);
    printf("cast operand: %.2f\n", (double)a / b);
    printf("cast result: %.2f\n", (double)(a / b));

    return 0;
}
````

### Obserwacja — ważne jest, **kiedy** wykonujemy konwersję

- `a / b` — oba argumenty to `int`, więc wynik wynosi **`0`**.
- `1.0 / 2` — literał `1.0` ma typ `double`; `2` zostaje **niejawnie przekonwertowane** do `double`, wynik to **`0.50`**.
- `(double)a / b` — najpierw rzutujemy **operand** `a` na `double`, więc dzielenie daje **`0.50`**.
- `(double)(a / b)` — najpierw obliczamy całkowite `a / b = 0`, a **dopiero potem** zmieniamy typ wyniku na `double`; otrzymujemy **`0.00`**.

**Pytanie:** czy końcowe rzutowanie może odzyskać część ułamkową, którą już utraciliśmy? **Nie.**

**Eksperyment:** zmień tylko `a = 1` na `a = 3`. Jakie wyniki otrzymasz?

### Przypisanie również może zmieniać typ wartości

W W2.3 używaliśmy przypisania `a = 10;`. Przy dwóch różnych typach C może wykonać **konwersję niejawną**.

Zobaczmy, co dzieje się, gdy przenosimy wartość **`double → int → double`**.

**Pytanie:** jeżeli zaczynamy od `6.28`, czy po dwóch przypisaniach znowu otrzymamy `6.28`?

````c
#include <stdio.h>

int main(void)
{
    double x = 6.28;
    int a = 2;

    printf("start: a=%d, x=%.2f\n", a, x);
    a = x;
    printf("after double to int: a=%d\n", a);
    x = a;
    printf("after int to double: x=%.2f\n", x);

    return 0;
}
````

### Obserwacja — czego nie da się odzyskać?

- `a = x;` — przypisujemy `double` do `int`. Dla wartości mieszczącej się w zakresie `int` **część ułamkowa zostaje odrzucona w kierunku zera**. `6.28` daje `6`.
- `x = a;` — konwersja z `int` do `double` tworzy wartość `6.0`. **Nie przywraca** utraconego `0.28`.

Wartość zmiennej `x` została zastąpiona nową wartością. Wcześniejszej liczby `6.28` nie przechowujemy już w `x`.

**Eksperyment:** zamień `double x = 6.28;` na `double x = -6.28;`. Czy `a` będzie równe `-6`, czy `-7`?

**Uwaga:** konwersja zmiennoprzecinkowa → całkowita poza reprezentowalnym zakresem typu docelowego jest w C *undefined behavior*. Takich wartości nie sprawdzamy w tym ćwiczeniu. Przy odpowiednich opcjach, np. GCC `-Wconversion`, kompilator może ostrzec o utracie danych.

### Konwersja jawna, czyli rzutowanie: `(int)x`

Możemy świadomie wskazać typ, na jaki konwertujemy wartość: **`(int)x`**.

Rzutowanie nie oznacza, że wartość zmiennej `x` zmienia swój typ. Tworzymy **wartość wynikową określonego typu**, którą możemy zapisać w innej zmiennej.

**Pytanie:** czy jawne rzutowanie `(int)x` zachowa więcej informacji niż wcześniejsze przypisanie `a = x`?

````c
#include <stdio.h>

int main(void)
{
    double x = 6.28;
    int a = (int)x;
    double y = (double)a;

    printf("original: %.2f\n", x);
    printf("cast to int: %d\n", a);
    printf("back to double: %.2f\n", y);

    return 0;
}
````

### Obserwacja — jawność nie usuwa utraty informacji

Wyniki to `6.28`, `6` i `6.00`.

- `(int)x` jawnie wyraża zamiar otrzymania wartości całkowitej. **Nie wykonuje zaokrąglenia do najbliższej liczby**, tylko odrzuca część ułamkową w kierunku zera (gdy konwersja jest zdefiniowana).
- `(double)a` daje wartość rzeczywistą `6.0`, ale **nie odtwarza `6.28`**.
- Samo rzutowanie nie zmienia typu **zmiennej**: `x` nadal jest `double` i zachowuje pierwotną wartość.

**Pytanie:** dlaczego wynik ostatniego `printf` to `6.00`, chociaż zmienna `y` ma typ `double`?

### Czy konwersja `int → float` zawsze jest dokładna?

Z W2.6 wiemy już, że `float` ma w konfiguracji odniesienia **24 bity cyfr znaczących**, a `double` ma **53 bity**.

**Pytanie:** jeśli `int` przechowuje dokładnie liczbę `16777217`, czy po przypisaniu jej do `float` nadal będziemy mogli odróżnić ją od `16777216`?

Sprawdźmy tę samą liczbę w trzech typach.

````c
#include <stdio.h>

int main(void)
{
    int a = 16777217;
    float f = a;
    double d = a;

    printf("int: %d\n", a);
    printf("float: %.0f\n", f);
    printf("double: %.0f\n", d);

    return 0;
}
````

### Obserwacja — duży `int` może stracić dokładność w `float`

W konfiguracji odniesienia zobaczymy:

```
int: 16777217
float: 16777216
double: 16777217
```

Liczba `16777217` mieści się w `int`, ale nie jest dokładnie reprezentowalna w 32-bitowym `float`. Przy stosowanym tu zaokrąglaniu IEEE 754 dostajemy **`16777216`**. `double` może przechować tę liczbę dokładnie.

**Wniosek:** nawet konwersja **`int → float`** może zgubić informację — mimo że typ `float` obsługuje części ułamkowe. Nie wystarczy wiedzieć, że typ jest „rzeczywisty”; trzeba pamiętać o **liczbie bitów precyzji**.

**Eksperyment:** zmień `16777217` na `100`. Czy wszystkie trzy wartości będą teraz identyczne?

### Uwaga na `printf`: format musi pasować do typu

**Rzutowanie wartości** i **sposób jej wyświetlenia** to dwie różne rzeczy.

- `%d` oczekuje argumentu typu `int`.
- `%f` (np. `%.2f`) oczekuje argumentu typu `double`. W wywołaniu `printf` wartości typu `float` są automatycznie przekazywane jako `double`.

**Antyprzykład — nie uruchamiaj:**

```c
double x = 6.28;
printf("%d\n", x);  // BŁĄD: niezgodność argumentu i formatu
```

To nie jest poprawna konwersja typu. Niezgodność specyfikatora z rzeczywistym typem argumentu może prowadzić do *undefined behavior*. Nie próbujemy zmieniać typu przez dobór `%d` lub `%f`.

### W2.7 — podsumowanie

- **Konwersja niejawna** następuje bez zapisu `(typ)` — np. w mieszanym wyrażeniu `1.0 / 2` lub przypisaniu `int → double`.
- **Konwersja jawna (rzutowanie)** ma składnię `(typ)wyrażenie` — np. `(double)a`.
- Miejsce konwersji ma znaczenie: **`(double)a / b`** różni się od **`(double)(a / b)`**.
- `double → int` może utracić część ułamkową, a `int → float` może utracić **precyzję**.
- Rzutowanie nie zmienia typu pierwotnej zmiennej ani nie odzyskuje raz utraconych danych.
- Format `printf` nie służy do konwersji wartości i musi odpowiadać rzeczywistemu typowi argumentu.

**Pytanie:** czy znak `'A'` jest w C przechowywany jako symbol, czy jako liczba? W W2.8 przyjrzymy się typowi `char` i kodowaniu znaków.

## W2.8 — `char`: jak komputer przechowuje znaki?

Do tej pory interpretowaliśmy wzorce bitów jako liczby. **Znaki** też są przechowywane jako wartości liczbowe — o tym, jaki znak reprezentuje dany kod, decyduje **kodowanie znaków**.

W konfiguracji odniesienia podstawowe znaki mają kody zgodne z **ASCII** (*American Standard Code for Information Interchange*, amerykański standard kodowania znaków). ASCII jest **7-bitowe** i obejmuje kody **0–127**. Przywracam również [**tablicę ASCII z oryginalnego wykładu**](https://www.asciitable.com/). Poniżej kilka przykładów; za chwilę zobaczymy pełną tablicę:

| Znak | Kod ASCII (dziesiętnie) | Kod binarny (8 bitów w konfiguracji odniesienia) |
|:-----|------------------------:|:---------------------------------|
| `'0'` | 48 | `00110000` |
| `'A'` | 65 | `01000001` |
| `'B'` | 66 | `01000010` |
| `'a'` | 97 | `01100001` |

W języku C typ **`char` zajmuje zawsze 1 bajt C**, czyli `sizeof(char) == 1`. W konfiguracji odniesienia `CHAR_BIT == 8`, więc mamy **8 bitów** w bajcie i mieści się w nim każdy kod ASCII. W innych implementacjach bajt C nie musi mieć dokładnie 8 bitów.

**Pytanie:** czy ta sama zmienna `char` może być wypisana jako **znak** i jako **jego kod liczbowy**?

**Pytanie:** znajdź kody `'0'`, `'A'`, `'a'` oraz `'\n'`. Które z nich odpowiadają widocznym znakom?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    char znak = 'A';

    printf("sizeof(char)=%zu B\n", sizeof(char));
    printf("CHAR_BIT=%d\n", CHAR_BIT);
    printf("symbol=%c\n", znak);
    printf("code=%d\n", znak);
    printf("'B' code=%d\n", 'B');
    return 0;
}
````

### Obserwacja — ten sam `char`, dwa sposoby wyświetlenia

W konfiguracji odniesienia otrzymujemy `symbol=A`, `code=65` i `'B' code=66`.

- `%c` w `printf` wyświetla **znak** odpowiadający kodowi.
- `%d` wyświetla **wartość liczbową** (przy przekazywaniu argumentu `char` następuje automatyczna promocja całkowita).
- `'A'` to **literał znakowy**, zapisany w pojedynczych apostrofach. W C zwykły literał znakowy, taki jak `'A'`, ma formalnie typ `int`, chociaż jego wartość może być przechowana w zmiennej `char`.

**Eksperyment:** zmień `'A'` na `'a'`. Przewidź, jaki znak i kod pojawią się w wyniku, a potem uruchom program.

**Ważne:** konkretne kody `65`, `66` czy `97` wynikają z ASCII na naszej platformie. Język C nie wymaga, by każda implementacja stosowała ASCII.

### `'0'`, `0` i `"0"` — trzy różne zapisy

W C zapis ma znaczenie:

- `'0'` — **literał znakowy** z kodem znaku cyfry zero (w konfiguracji odniesienia: 48).
- `0` — **liczba całkowita zero**, a nie znak cyfry.
- `"0"` — **napis**, czyli sekwencja znaków zakończona specjalnym znakiem **`'\0'`** (*NUL*, kod 0).

Napis `"0"` zajmuje więc **2 bajty C**: kod cyfry `'0'` oraz kończący `'\0'`. Napis wyświetlamy przez `%s`, a pojedynczy znak przez `%c`.

**Pytanie:** co zwróci `sizeof("0")` — 1 czy 2?

````c
#include <stdio.h>

int main(void)
{
    char cyfra = '0';
    char zero = 0;

    printf("character: %c, code: %d\n", cyfra, cyfra);
    printf("numeric zero: %d\n", zero);
    printf("string: %s\n", "0");
    printf("string bytes: %zu\n", sizeof("0"));
    printf("NUL code: %d\n", '\0');
    printf("newline code: %d\n", '\n');
    return 0;
}
````

### Obserwacja — ASCII zawiera także znaki sterujące

W konfiguracji odniesienia program wypisuje `character: 0, code: 48`, `numeric zero: 0` i `string bytes: 2`.

Znak **`'\0'`** (*NUL*) ma kod **0**; w C oznacza m.in. **koniec napisu**. Jest to znak o kodzie zero, nie litera `'0'` (kod 48). Nie mylmy go także z makrem **`NULL`**, dotyczącym wskaźników, które poznamy później.

Zapis **`'\n'`** oznacza pojedynczy znak **nowego wiersza**, o kodzie ASCII **10** w konfiguracji odniesienia. Takie znaki sterujące najlepiej najpierw pokazać jako **wartości liczbowe**, ponieważ ich wyświetlenie przez `%c` nie musi dawać widocznego symbolu.

**Eksperyment:** zamień `char cyfra = '0';` na `char cyfra = '7';`. Czy zmienia się jedynie widoczny symbol, czy także jego kod liczbowy?

### `signed char`, `unsigned char` i zakres przechowywanych wartości

Typ `char` przechowuje **wartość liczbową**, a jego reprezentacja nie zawsze ogranicza się do kodów ASCII. Możemy jawnie użyć:

- **`signed char`** — liczb ze znakiem;
- **`unsigned char`** — liczb bez znaku.

Zwykły **`char` jest odrębnym typem**, a to, czy zachowuje się w zakresie jak `signed char` czy `unsigned char`, zależy od **implementacji C**. Sprawdźmy to, zamiast zakładać, że `char a = 128;` zawsze da taki sam wynik.

**Pytanie:** co oznacza wzorzec `11111111` dla `signed char` i `unsigned char` na naszym 8-bitowym komputerze?

````c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    signed char a = -1;
    unsigned char b = UCHAR_MAX;

    printf("CHAR_MIN=%d CHAR_MAX=%d\n", CHAR_MIN, CHAR_MAX);
    printf("SCHAR_MIN=%d SCHAR_MAX=%d\n", SCHAR_MIN, SCHAR_MAX);
    printf("UCHAR_MAX=%u\n", (unsigned int)UCHAR_MAX);
    printf("signed char: %d\n", a);
    printf("unsigned char: %u\n", (unsigned int)b);
    return 0;
}
````

### Obserwacja — te same bity, dwa zakresy

W konfiguracji odniesienia bajt ma 8 bitów i zwykły `char` jest typu ze znakiem. Otrzymujemy:

- `CHAR_MIN = SCHAR_MIN = −128`, a `CHAR_MAX = SCHAR_MAX = 127`;
- `UCHAR_MAX = 255`;
- `signed char` może reprezentować `−1` jako `11111111`;
- `unsigned char` interpretuje `11111111` jako `255`.

To nawiązanie do reprezentacji `int` i `unsigned int` poznanej w W2.4–W2.5. **Nie wolno zakładać**, że przypisanie liczby `128` do zwykłego `char` da `−128` na każdej platformie — gdy wartość nie mieści się w typie ze znakiem, wynik konwersji całkowitej jest zależny od implementacji (albo może zostać zgłoszony sygnał zależny od implementacji).

W `printf` jawnie używamy `(unsigned int)` przy `%u`, żeby argument miał odpowiedni typ. Tę składnię rzutowania poznaliśmy już w W2.7.

**Pytanie:** czy wszystkie wartości od 128 do 255 są znakami ASCII? **Nie.** ASCII opisuje zakres 0–127; znaczenie pozostałych bajtów zależy od zastosowanego kodowania.

### W2.8 — podsumowanie

- **`char` zajmuje jeden bajt C**, a liczba bitów w tym bajcie wynosi `CHAR_BIT` (w konfiguracji odniesienia: 8).
- Kodowanie **ASCII** przypisuje znakom kody 0–127; w szczególności `'A'` ma kod 65, a `'0'` — 48.
- `%c` wyświetla znak, a `%d` — jego wartość liczbową.
- **`'0'`**, **`0`** i **`"0"`** nie oznaczają tego samego.
- Zwykły `char` może mieć zakres ze znakiem lub bez znaku, zależnie od implementacji.

**Pytanie:** jak w C zapisać wartość, która może być tylko **prawdziwa albo fałszywa**? W W2.9 przyjrzymy się typowi logicznemu `bool`.

Rozwinięte kodowania tekstu, takie jak **UTF-8 (Unicode Transformation Format, 8-bitowa postać kodowania)**, poznamy w późniejszym materiale. Tutaj ograniczamy się do podstawowych znaków ASCII.

## W2.9 — `bool`: prawda i fałsz

Czasami wynik pytania może być tylko **prawdziwy** albo **fałszywy**, np. „Czy `a` jest mniejsze od `b`?”. W C możemy przechowywać taką informację w zmiennej typu **`bool`**.

Od standardu **C99** język ma wbudowany typ **`_Bool`**. W używanym przez nas C11 nagłówek **`<stdbool.h>`** udostępnia wygodne nazwy:

- `bool` — nazwa typu logicznego (w C11 odpowiada `_Bool`);
- `true` — prawda, wartość logiczna **1**;
- `false` — fałsz, wartość logiczna **0**.

Do wyświetlenia liczbowej wartości `bool` w `printf` użyjemy `%d`, ponieważ argument `bool` zostaje przy takim wywołaniu przekonwertowany na `int`.

**Pytanie:** skoro wartości są tylko dwie, czy `bool` musi zajmować dokładnie jeden **bit** w pamięci? Sprawdźmy `sizeof(bool)` — wynik `sizeof` zawsze podaje liczbę **bajtów C**.

````c
#include <stdio.h>
#include <stdbool.h>

int main(void)
{
    bool gotowe = true;

    printf("sizeof(bool)=%zu B\n", sizeof(bool));
    printf("gotowe=%d\n", gotowe);
    gotowe = false;
    printf("po zmianie=%d\n", gotowe);
    return 0;
}
````

### Obserwacja — dwie wartości, ale niekoniecznie jeden bit pamięci

W konfiguracji odniesienia program wypisuje:

```
sizeof(bool)=1 B
gotowe=1
po zmianie=0
```

**Ważne:** `bool` może przyjmować tylko **0** i **1**, ale `sizeof(bool)` mierzy zajęte **bajty**, nie liczbę możliwych wartości. W konfiguracji odniesienia taki obiekt ma **1 bajt (8 bitów)**; standard C nie nakazuje jednak, aby wszędzie zajmował dokładnie jeden bajt.

Dla porównania z W2.4, na naszej platformie typowa reprezentacja tych wartości wygląda tak:

| Wartość logiczna | Wartość liczbowa | Typowy zapis w 8 bitach |
|:-----------------|-----------------:|:------------------------|
| `false` | 0 | `00000000` |
| `true` | 1 | `00000001` |

To ilustracja dla naszej implementacji, **nie** uniwersalna gwarancja układu bitów każdego obiektu `bool`.

**Eksperyment:** zmień wartość początkową `gotowe` z `true` na `false`. Czy `sizeof(bool)` się zmieni?

### Co się stanie, gdy do `bool` przypiszemy zwykłą liczbę?

Typ logiczny przechowuje tylko wartości **0** lub **1**. Przy konwersji liczby całkowitej na `bool`:

- `0` staje się **`false` (0)**;
- **każda wartość niezerowa** staje się **`true` (1)**.

Dotyczy to także liczb ujemnych.

**Pytanie:** czy po przypisaniu `42` do `bool` program wypisze `42`, czy `1`?

````c
#include <stdio.h>
#include <stdbool.h>

int main(void)
{
    bool a = 0;
    bool b = 42;
    bool c = -3;

    printf("0 -> %d\n", a);
    printf("42 -> %d\n", b);
    printf("-3 -> %d\n", c);
    return 0;
}
````

### Obserwacja — `bool` nie przechowuje dowolnej liczby

Wyniki to `0 → 0`, `42 → 1` oraz `−3 → 1`. W drugim przypadku w zmiennej `b` **nie została zapamiętana liczba 42**: nastąpiła konwersja na wartość logiczną `1`.

Skąd biorą się wartości prawda/fałsz w programach? Między innymi z **porównań**:

- `a < b` — czy `a` jest mniejsze od `b`?
- `a == b` — czy wartości są **równe**?
- `a != b` — czy wartości są **różne**?

**Uwaga na `=` i `==`:** pojedyncze `=` to **przypisanie**, natomiast podwójne `==` to **porównanie**.

W języku C wynik takiego porównania ma formalnie **typ `int`**, o wartości `0` (fałsz) albo `1` (prawda). Możemy ten wynik przypisać do `bool`.

**Pytanie:** jakie trzy wyniki otrzymamy dla `a = 3` i `b = 5`?

````c
#include <stdio.h>
#include <stdbool.h>

int main(void)
{
    int a = 3, b = 5;
    bool mniejszy = (a < b);
    bool rowne = (a == b);
    bool rozne = (a != b);

    printf("a<b=%d\n", mniejszy);
    printf("a==b=%d\n", rowne);
    printf("a!=b=%d\n", rozne);
    return 0;
}
````

### W2.9 — podsumowanie

- `bool` przechowuje **fałsz (0)** albo **prawdę (1)**; w C11 korzystamy z `<stdbool.h>`.
- W konfiguracji odniesienia `sizeof(bool) = 1` bajt C, ale nie oznacza to, że `bool` zajmuje tylko jeden bit pamięci.
- Po przypisaniu **zera** dostajemy `false`; po przypisaniu **wartości niezerowej** dostajemy `true`.
- Porównania `<`, `==` i `!=` dają wartości 0 albo 1; w C wynik porównania ma formalnie typ **`int`**, ale można zapisać go w `bool`.
- **`=` przypisuje**, natomiast **`==` porównuje**.

**Eksperyment:** zmień `b = 5` na `b = 3` w ostatnim programie. Co zmieni się w wynikach trzech porównań?

Na razie badamy same wartości logiczne — do instrukcji warunkowych wrócimy później. **Pytanie:** jak w C opisać funkcję, która nie zwraca żadnej wartości? W W2.10 wrócimy do `void`, znanego już z W1.

## W2.10 — `void` i funkcje: wracamy do W1

W pierwszym wykładzie używaliśmy funkcji `sum` i `main`, zanim znaliśmy wszystkie szczegóły typów danych. Wróćmy do nich teraz, po poznaniu `int`, `double`, `char` i `bool`.

**`void` oznacza brak wartości danego typu.** Zobaczmy dwa miejsca, gdzie pojawia się w zapisie funkcji:

```c
void przywitaj(void)
```

- **Pierwsze `void`**: funkcja **nie zwraca wartości** wywołującemu kodowi.
- **`(void)`**: funkcja **nie przyjmuje żadnych argumentów**.

`void` jest **typem**, ale nie typem zwykłej zmiennej, do której wpisujemy dane.

**Pytanie:** czy funkcja, która niczego nie zwraca, może wykonać pożyteczną pracę?

````c
#include <stdio.h>

void przywitaj(void); // deklaracja / prototyp

int main(void)
{
    printf("main: poczatek\n");
    przywitaj();
    printf("main: koniec\n");
    return 0;
}

void przywitaj(void) // definicja funkcji
{
    printf("Witaj z funkcji przywitaj\n");
    return;
}
````

### Obserwacja — brak wyniku nie oznacza braku działania

Funkcja `przywitaj` wypisuje tekst, ale **nie przekazuje żadnej wartości** do `main`.

- `void przywitaj(void);` to **deklaracja (prototyp)**. Kompilator poznaje interfejs funkcji przed jej wywołaniem.
- `void przywitaj(void) { ... }` to **definicja**, zawierająca instrukcje do wykonania.
- `return;` w funkcji `void` **kończy jej wykonanie bez zwracania wartości**. W tym miejscu nie jest konieczne, bo funkcja i tak kończy się na klamrze `}`.

**Pytanie:** jak zmieni się wynik programu, jeżeli usuniemy ostatnią instrukcję `return;` z funkcji `przywitaj`? Czy trzeba zmienić `return 0;` w `main`?

**Ważne:** w zwykłym programie C funkcja `main` ma **zwracać `int`**. Nie zamieniamy jej na `void main(void)`. Zapis `int main(void)` i `return 0;` pozostają poprawne.

### `void` kontra `int`: wypisanie wyniku a zwrócenie wyniku

W W1 zaczynaliśmy od funkcji, która **wypisywała** sumę. Dopiero później napisaliśmy funkcję, która **zwracała** wartość.

Porównajmy je w jednym programie:

- `void pokaz_sume(int a, int b)` — wykonuje działanie i **wypisuje** wynik.
- `int policz_sume(int a, int b)` — oblicza i **zwraca** wynik do miejsca wywołania.

**Pytanie:** która z tych funkcji pozwoli łatwo zapisać obliczony wynik w zmiennej `wynik`?

````c
#include <stdio.h>

void pokaz_sume(int a, int b)
{
    printf("void: %d\n", a + b);
}

int policz_sume(int a, int b)
{
    return a + b;
}

int main(void)
{
    pokaz_sume(5, 6);
    int wynik = policz_sume(5, 6);
    printf("int: %d\n", wynik);
    return 0;
}
````

## W2.11 — Operatory: priorytet, inkrementacja i przypisanie złożone

Znamy już typy danych i podstawowe działania. Teraz sprawdzimy, **jak C interpretuje wyrażenia, gdy występuje w nich kilka operatorów**.

Priorytet operatorów (*precedence*) określa, **jak wyrażenie jest grupowane**, a nie to, w jakiej kolejności procesor musi obliczać poszczególne fragmenty. Nawiasy pozwalają nam jawnie wskazać grupowanie.

| Operatory | Co oznaczają? | Priorytet wśród poznanych działań |
|:----------|:--------------|:----------------------------------|
| `( )` | Jawne grupowanie wyrażenia | Zmienia domyślne grupowanie |
| `*`, `/`, `%` | Mnożenie, dzielenie, reszta | Wyższy niż `+` i `−` |
| `+`, `-` | Dodawanie i odejmowanie | Niższy niż `*` i `/` |
| `<`, `==` | Porównania z W2.9 | Niższy niż działania arytmetyczne |
| `=`, `+=` itd. | Przypisania | Niższy niż powyższe operatory |

Pełnej tabeli operatorów nie musimy teraz zapamiętywać. Zwracajmy uwagę na te, które właśnie wykorzystujemy.

**Pytanie:** czy `2 + 3 * 4` daje `20`, czy `14`? Co zmienią nawiasy?

````c
#include <stdio.h>

int main(void)
{
    int a = 2, b = 3, c = 4;

    printf("a+b*c=%d\n", a + b * c);
    printf("(a+b)*c=%d\n", (a + b) * c);
    printf("20/5*2=%d\n", 20 / 5 * 2);
    printf("20/(5*2)=%d\n", 20 / (5 * 2));
    printf("a+b*c==14: %d\n", a + b * c == 14);

    return 0;
}
````

### Obserwacja — grupowanie działań

- `a + b * c = 14`, bo mnożenie wiąże silniej niż dodawanie: `a + (b * c)`.
- `(a + b) * c = 20`, bo nawias zmienia grupowanie.
- `20 / 5 * 2 = 8`. Operatory `/` i `*` mają **ten sam priorytet** i **łączność lewostronną**: wyrażenie jest grupowane jako `(20 / 5) * 2`.
- `20 / (5 * 2) = 2`, bo nawias nakazuje inne grupowanie.
- `a + b * c == 14` daje `1`. Najpierw ustalamy wartość całego wyrażenia arytmetycznego, a potem wynik porównania.

**Ważne:** priorytet i łączność operatorów decydują o **strukturze wyrażenia**. Nie oznacza to ogólnej obietnicy, że wszystkie jego fragmenty będą wykonywane od lewej do prawej.

**Eksperyment:** zamień `c = 4` na `c = 5`. Zanim uruchomisz program, przewidź dwie pierwsze wartości oraz wynik porównania.

Porównajmy dwa wyrażenia z liczbami typu `double`:

W pierwszym przypadku wyrażenie `b / (a * b)` ma **mianownik z nawiasami**. W drugim przypadku `b / a * b` oznacza `(b / a) * b`, ponieważ `/` i `*` mają ten sam priorytet i łączność lewostronną.

**Pytanie:** czy ta różnica ma znaczenie, gdy `a = 6.0` i `b = 9.0`?

````c
#include <stdio.h>

int main(void)
{
    double a = 6.0, b = 9.0;

    double c = (a + b) / a + b / (a * b);
    printf("z nawiasami=%.6f\n", c);

    c = (a + b) / a + b / a * b;
    printf("bez nawiasow=%.6f\n", c);

    return 0;
}
````


### `++` i `--`: zwiększanie i zmniejszanie o jeden

W C możemy zmienić wartość zmiennej o jeden, używając:

- `++a` — inkrementacja **prefiksowa**;
- `a++` — inkrementacja **postfiksowa**;
- `--a` — dekrementacja prefiksowa;
- `a--` — dekrementacja postfiksowa.

Gdy operator stanowi **osobną instrukcję**, oba warianty inkrementacji prowadzą do tej samej nowej wartości zmiennej. Podobnie zachowują się dwa warianty dekrementacji.

**Pytanie:** jakie cztery wartości zostaną wypisane, gdy zaczniemy od `a = 2`?

````c
#include <stdio.h>

int main(void)
{
    int a = 2;

    a++;
    printf("a++: %d\n", a);
    ++a;
    printf("++a: %d\n", a);
    a--;
    printf("a--: %d\n", a);
    --a;
    printf("--a: %d\n", a);

    return 0;
}
````

### Obserwacja — samodzielne `++a` i `a++` zmieniają `a` tak samo

Widzimy kolejno: **3, 4, 3, 2**. W każdym przypadku wartość zmiennej zmienia się o jeden.

**Eksperyment:** zamień początkowe `a = 2` na `a = 10`. Przewidź cztery wyniki.

**Pytanie:** po co więc istnieją dwie postacie, `++a` i `a++`? Różnica pojawia się wtedy, gdy **wartość wyrażenia** wykorzystujemy w innym miejscu, np. przy przypisaniu do drugiej zmiennej.

### Prefiks i postfiks: co zostaje przypisane do `b`?

Dla `int b = ++a;` następuje zwiększenie `a`, a **wartością wyrażenia** jest już wartość po zwiększeniu.

Dla `int b = a++;` **wartością wyrażenia** jest poprzednia wartość `a`, chociaż sama zmienna również zostaje zwiększona.

**Pytanie:** gdy `a = 1`, jakie wartości będą miały `a` i `b` po każdej z tych instrukcji?

````c
#include <stdio.h>

int main(void)
{
    int a = 1;
    int b = ++a;
    printf("prefix: a=%d b=%d\n", a, b);

    a = 1;
    b = a++;
    printf("postfix: a=%d b=%d\n", a, b);

    return 0;
}
````

### Obserwacja — wartość wyrażenia jest inna

```
prefix: a=2 b=2
postfix: a=2 b=1
```

W **obu przypadkach końcowe `a` wynosi 2**, ale `b` otrzymuje inną wartość. To jest rzeczywista różnica pomiędzy wersją prefiksową i postfiksową.

W historycznym materiale sugerowaliśmy, że prefiks jest automatycznie szybszy, bo postfiks wymaga kopiowania. **Dla zwykłych typów liczbowych w C nie jest to prawidłowa reguła wydajnościowa** — kompilator może wytworzyć równie wydajny kod dla obu wersji.

**Uwaga:** nie należy modyfikować tej samej zmiennej kilka razy w jednym wyrażeniu bez zapewnionej kolejności. Na przykład `int b = a++ + a++;` ma w C **niezdefiniowane zachowanie**. **Nie uruchamiamy** takiego kodu jako przykładu poprawnej inkrementacji. Dla początkujących bezpieczniej jest rozdzielać modyfikacje na osobne instrukcje.

### Operatory przypisania złożonego: `+=`, `-=`, `*=`, `/=`, `%=`

C pozwala zapisać częste operacje krócej:

| Zapis | Dla prostych zmiennych `int` odpowiada działaniu |
|:------|:----------------------------------------------|
| `a += b;` | `a = a + b;` |
| `a -= b;` | `a = a - b;` |
| `a *= b;` | `a = a * b;` |
| `a /= b;` | `a = a / b;` |
| `a %= b;` | `a = a % b;` |

To **operatory języka C**, nie skróty wprowadzane przez edytor. Dla zwykłych zmiennych `int` w tym ćwiczeniu zachowują się tak jak pokazane działania.

**Pytanie:** jeśli zaczniemy od `a = 10`, `b = 3`, co zostanie po pięciu kolejnych modyfikacjach `a`?

````c
#include <stdio.h>

int main(void)
{
    int a = 10, b = 3;

    a += b;
    printf("a+=b: %d\n", a);
    a -= b;
    printf("a-=b: %d\n", a);
    a *= b;
    printf("a*=b: %d\n", a);
    a /= b;
    printf("a/=b: %d\n", a);
    a %= b;
    printf("a%%=b: %d\n", a);

    return 0;
}
````

### Obserwacja — zmieniamy tę samą wartość krok po kroku

```
a+=b: 13
a-=b: 10
a*=b: 30
a/=b: 10
a%=b: 1
```

Zwróć uwagę, że wyniki każdej instrukcji wpływają na kolejną. Zapis `a %= b` korzysta z reszty z dzielenia poznanej w W2.5. Dla `printf` piszemy `%%`, żeby wypisać sam znak procenta.

Operatory `+=`, `-=`, `*=` i `/=` mogą również działać na `float` i `double`. **Operator `%=` jest przeznaczony do typów całkowitych**. Przy dzieleniu nadal obowiązuje zakaz dzielenia przez zero, a operacje na typie `int` nadal muszą respektować jego zakres.

- `++a` i `a++` osobno zwiększają `a` tak samo, ale mają **inną wartość jako wyrażenia**.

To **declare** a function in **C** we need to:
* specify its `type` (`int`, `double`, `void` ...)
* give it a name, an identifier
* decide how many and what arguments it need.

It could look like this:  
`type identifier(type arg1, type arg2)`  
We can define the **body** of the function just after the declaration, or postpone and define it elsewhere. The type, identifier and type, order and number of arguments must be the same in the declaration and definition.  
Examples:

````c
#include <stdio.h>

//this is a new function that we just added
void sum() // definicja funkcji
{
    printf("This is a printf from function sum()\n");// we need to accept that %d must be there
}

int main()
{
    sum(); // wywolanie funkcji
}
````

````c
#include <stdio.h>

void sum(); // deklaracja funkcji

int main()
{
    sum(); // wywolanie funkcji
}

void sum() // definicja funkcji
{
    printf("This is a printf from function sum()\n");// we need to accept that %d must be there
}
````

The type of the function determines its ability to *produce* a value, or to `return`. The `return` statement terminates a function, returns control back to the caller and also passes the **value** to the calling function.
*Note: the value passed by the `return` is not the same variable, just the value is the same.*

````c
#include <stdio.h>

void fun()
{
//     return;
    printf("Hello \n");
}
int fun2()
{
    return 6;
}
double fun3()
{
    double a;
    return 3.141592;
}

int main()
{
    int a =0;
    fun();
//     a = fun();
    printf("%d \n", a);
    a = fun2();
    printf("%d \n", a);
    
    double b =0;
    printf("%lf \n", b);
    b = fun3();
    printf("%lf \n", b);
    
    a = printf("%d \n", a);
    printf("%d \n", a);
}
````
